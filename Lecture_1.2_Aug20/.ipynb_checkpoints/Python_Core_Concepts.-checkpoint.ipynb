## 1. Constants & Variables: The Foundation of Program State

### 🎯 Conceptual Understanding:
* **Variables (Dynamic Labels):** In Python, a variable is not a physical memory container (box); it is a **name tag (pointer)** attached to an object created in heap memory.
* **Constants (Fixed Values):** Python has no native `const` keyword. By convention (PEP 8), variables intended to be constant are written in **`ALL_CAPS_WITH_UNDERSCORES`** (e.g., `SPEED_OF_LIGHT = 299792458`). In modern Python (3.8+), the `typing.Final` annotation can be used to declare constants statically for linters.

### 📋 Rules for Legal Variable Names:
1. Must begin with a letter (`a-z`, `A-Z`) or an underscore (`_`).
2. Cannot begin with a number (e.g., `2nd_sensor` is **illegal**; `sensor_2` is **legal**).
3. Can only contain alphanumeric characters and underscores (`A-z`, `0-9`, and `_`).
4. Case-sensitive: `learning_rate`, `Learning_Rate`, and `LEARNING_RATE` are three distinct entities.
5. Cannot use Python reserved keywords (e.g., `for`, `class`, `import`, `def`, `lambda`, `pass`).

````python
# Real-World Scenario: Atmospheric Physics Constants & Sensor Variables
from typing import Final

# 1. Defining Constants ( search for PEP 8 Convention + Typing Annotation  on web for  more info)

STANDARD_GRAVITY: Final[float] = 9.80665      
SEA_LEVEL_PRESSURE: Final[float] = 1013.25    
BOLTZMANN_CONSTANT: Final[float] = 1.380649e-23 

######
# Caution  constants like this are available only from python 3.8 and above.
######

# 2. Defining Dynamic Sensor Variables
sensor_station_id = "ICFOSS-WEATHER-NODE-01"
current_altitude_m = 45.8
is_active = True

print(f"Station       : {sensor_station_id}")    # f will be explained later
print(f"Gravity Const : {STANDARD_GRAVITY} m/s²")
print(f"Altitude      : {current_altitude_m} meters")
print(f"Active State  : {is_active}")
````

## 2. Core Primitive Data Types & Type Casting

Python is **dynamically typed** (variable types are inferred at runtime) and **strongly typed** (Python will not silently convert incompatible types without explicit instruction).

### 📊 The Fundamental Type Hierarchy:
| Type | Python Class | Characteristics | Real-World Example |
| :--- | :--- | :--- | :--- |
| **Integer** | `int` | Unbounded arbitrary-precision whole numbers | `sample_count = 10000` |
| **Floating-Point** | `float` | 64-bit double precision (IEEE-754) | `voltage = 3.304` |
| **Complex** | `complex` | Number with `.real` and `.imag` ($j = \sqrt{-1}$) | `impedance = 50 + 12.5j` |
| **String** | `str` | Immutable Unicode character sequence | `location = "Thiruvananthapuram"` |
| **Boolean** | `bool` | Subclass of `int` (`True == 1`, `False == 0`) | `network_connected = True` |
| **NoneType** | `NoneType` | Singleton representing absence of value (`None`) | `error_code = None` |

### 🔄 Explicit Type Casting:
* `int(x)`: Converts to integer (truncates floats towards zero; parses string integers).
* `float(x)`: Converts integers or valid numeric strings to floats.
* `str(x)`: Converts any object into its string representation.
* `bool(x)`: Evaluates truthiness (`0`, `0.0`, `""`, `None`, `[]` become `False`; everything else becomes `True`).

````python
# Demonstrating Dynamic Typing, Type Inspection and Type Casting
raw_input_str = "4250"
raw_voltage_str = "3.285"

# 1. Explicit Conversion
parsed_int = int(raw_input_str)
parsed_float = float(raw_voltage_str)

print("Type of raw_input_str :", type(raw_input_str))
print("Type of parsed_int    :", type(parsed_int), f"| Value + 50 = {parsed_int + 50}")
print("Type of parsed_float  :", type(parsed_float), f"| Value * 2 = {parsed_float * 2:.3f}")

# 2. Safe Type Verification with isinstance()
print("\nisinstance checks:")
print("is parsed_int an int?   :", isinstance(parsed_int, int))
print("is True an instance of int?:", isinstance(True, int))  # True is an int in Python!

# 3. Complex numbers
z = 4 + 3j
print(f"\nComplex number: {z} | Real part: {z.real} | Imag part: {z.imag} | Magnitude: {abs(z)}")
````

## 3. Expressions vs. Statements

A fundamental distinction in computer science:

### 🔹 What is an Expression?
An **expression** is any valid unit of code that **evaluates to a value**.
* Examples: `5 + 3`, `math.sqrt(16)`, `x > 10`, `"ICFOSS " * 2`.
* Expressions can appear anywhere a value is expected (e.g., as function arguments or right-hand side of assignments).

### 🔹 What is a Statement?
A **statement** is an instruction that **performs an action** or controls program execution. Statements do not return values directly.
* **Assignment Statement:** `x = 100`
* **Import Statement:** `import math`
* **Control Flow Statements:** `if ... else`, `for`, `while`
* **Null Operation Statement:** `pass`

````python
# Expressions evaluated inside statements
import math  # Statement: Import

radius = 5.0 # Statement: Assignment (where 5.0 is a literal expression)

# The right side is a compound expression combining operators and function calls:
circle_area = math.pi * (radius ** 2)

print(f"Radius: {radius} cm -> Area: {circle_area:.2f} cm²")
````

## 4. Python Operators in Depth

Operators are special symbols that carry out computation on operands.

### 4.1 Arithmetic Operators & Division Mechanics

| Operator | Operation | Description | Example ($a=17, b=5$) |
| :---: | :--- | :--- | :--- |
| `+` | Addition | Sum of operands | `17 + 5 = 22` |
| `-` | Subtraction | Difference of operands | `17 - 5 = 12` |
| `*` | Multiplication | Product of operands | `17 * 5 = 85` |
| `/` | **True Division** | Always produces a floating-point quotient | `17 / 5 = 3.4` |
| `//` | **Floor Division** | Quotient rounded towards $-\infty$ (mathematical floor) | `17 // 5 = 3`, `-17 // 5 = -4` |
| `%` | **Modulo** | Remainder after division | `17 % 5 = 2`, `-17 % 5 = 3` |
| `**` | Exponentiation | Left operand raised to power of right | `2 ** 8 = 256` |

````python
# Arithmetic Demonstrations
a = 17
b = 5

print(f"True Division  (17 /  5) = {a / b}")
print(f"Floor Division (17 // 5) = {a // b}")
print(f"Modulo         (17 %  5) = {a % b}")

# Negative numbers and the Python Modulo Invariant
# Invariant: a == (a // b) * b + (a % b)
neg_a = -17
print(f"\nNegative Floor (-17 // 5) = {neg_a // b}  (Rounds down towards -infinity)")
print(f"Negative Modulo (-17 %  5) = {neg_a % b}")
print(f"Invariant verified: {neg_a == (neg_a // b) * b + (neg_a % b)}")
````

### 4.2 Comparison, Logical & Identity Operators

| Category | Operators | Key Characteristics |
| :--- | :--- | :--- |
| **Relational / Comparison** | `==`, `!=`, `<`, `>`, `<=`, `>=` | Compare values, return `True` or `False`. Supports mathematical chaining (`10 < x <= 20`). |
| **Logical** | `and`, `or`, `not` | Evaluate boolean conditions using **short-circuit evaluation**. Return deciding operand. |
| **Identity** | `is`, `is not` | Test whether two variables point to the **exact same memory address** (`id(a) == id(b)`). |
| **Membership** | `in`, `not in` | Test whether an element exists inside a collection, sequence, or string. |

````python
# 1. Chained Comparisons
speed = 65
print(f"Is speed within highway limits (50 <= speed <= 80)? {50 <= speed <= 80}")

# 2. Short-Circuit Logic (and / or return the deciding operand)
user_setting = None
default_setting = "DEFAULT_ICFOSS_PROFILE"
active_setting = user_setting or default_setting
print(f"Active Setting: {active_setting}")

# 3. Membership Testing
allowed_roles = ["admin", "researcher", "student"]
current_user_role = "researcher"
print(f"Access Granted? {current_user_role in allowed_roles}")

# 4. Identity (is) vs Equality (==)
list_1 = [1, 2, 3]
list_2 = [1, 2, 3]
print(f"list_1 == list_2 (Values match): {list_1 == list_2}")
print(f"list_1 is list_2 (Same memory) : {list_1 is list_2} (id1: {id(list_1)}, id2: {id(list_2)})")
````

## 5. Running Python Scripts from the Command Line

While Jupyter Notebooks are ideal for interactive experimentation, production AI systems and automation pipelines are deployed as standalone Python scripts (`.py`).

### 🖥️ Two Modes of Execution:

**1. Interactive Mode (REPL):** Type `python3` in your terminal to get the `>>>` prompt. Each statement is evaluated and echoed immediately. Excellent for quick checks and exploration, but nothing is saved or reproducible.

**2. Script Mode:** Hand the interpreter a `.py` file. It executes the file top to bottom and exits.
* **Without arguments:** `python3 my_script.py` — all behaviour is fixed inside the file.
* **With arguments:** `python3 process_data.py --dataset sales.csv --epochs 50` — the *same* execution mode, except the values are supplied at runtime instead of being hard-coded. This is what turns a one-off script into a reusable tool.

> ⚠️ **Teaching note:** Passing command-line arguments is **not** a third way of running Python. The interpreter starts in exactly the same manner; only the contents of `sys.argv` differ. The real distinction is *interactive vs. script*, and arguments are simply how a script is parameterised.

*(Two further invocation forms worth knowing, both still non-interactive: `python3 -c "print(2**10)"` runs a code string directly, and `python3 -m http.server` runs an installed module as a script.)*

### 📥 Accessing Command-Line Arguments via `sys.argv`:
* `sys.argv` is a **list of strings** — numeric arguments must be cast explicitly with `int()` or `float()`.
* `sys.argv[0]` is always the script name itself.
* `sys.argv[1]`, `sys.argv[2]`, ... contain the arguments passed after the script name.
* `len(sys.argv)` tells you how many were supplied — always validate this before indexing, or a missing argument raises `IndexError`.

Let's write a standalone script to disk using Jupyter's `%%writefile` magic and execute it directly!

````python
%%writefile score_report.py
# score_report.py - A simple standalone command-line script
import sys

# 1. Constant (Section 1)
PASS_PERCENTAGE = 40.0

print("=" * 45)
print("  ICFOSS SCORE REPORT GENERATOR")
print("=" * 45)

# 2. sys.argv is simply a LIST OF STRINGS
print(f"Script name     : {sys.argv[0]}")
print(f"Arguments given : {len(sys.argv) - 1}")

# 3. Always check how many arguments arrived BEFORE indexing the list
if len(sys.argv) < 4:
    print("\nUsage  : python score_report.py <student_name> <marks_scored> <total_marks>")
    print("Example: python score_report.py Anjali 68 80")
else:
    # 4. Type casting (Section 2): arguments arrive as text, never as numbers
    student_name = sys.argv[1]
    marks_scored = float(sys.argv[2])
    total_marks  = float(sys.argv[3])

    # 5. Expressions and operators (Sections 3 & 4)
    percentage = (marks_scored / total_marks) * 100
    has_passed = percentage >= PASS_PERCENTAGE   # comparison returns a bool

    print(f"\nStudent    : {student_name}")
    print(f"Score      : {marks_scored} out of {total_marks}")
    print(f"Percentage : {percentage:.2f}%")
    print(f"Passed     : {has_passed}")
````

### 🚀 Executing the Script from the Command Line:
Inside a notebook we can invoke the script with the shell prefix `!`. In a real terminal you would drop the `!` and type the command directly.

Run it twice — once correctly, and once with an argument missing — to see why the `len(sys.argv)` check matters:


````python
# 1. Correct usage - all three arguments supplied
!python score_report.py Anjali 68 80

# 2. Missing arguments - the script prints its usage message instead of crashing
!python score_report.py Anjali
````

## 6. Importing Libraries & Modules

A **module** is simply a Python file (`.py`) containing functions, classes, and variables. A **package** is a directory of modules. Importing is how we reuse code that lives outside the current file.

### 📦 Four Forms of the `import` Statement:
| Form | Example | What name(s) you get | How you call it |
| :--- | :--- | :--- | :--- |
| Plain import | `import math` | `math` (the module object) | `math.sqrt(25)` |
| Aliased import | `import numpy as np` | `np` | `np.array([1, 2])` |
| From-import | `from math import sqrt, pi` | `sqrt`, `pi` | `sqrt(25)` |
| Star import ⚠️ | `from math import *` | *every* public name in the module | `sqrt(25)` |

### 🔑 `import X` vs. `from X import y` — the Real Difference

This is the single most misunderstood point in this section, so let's be precise.

**Both statements do exactly the same work behind the scenes.** In both cases Python locates the module, executes it **completely**, and stores it in the `sys.modules` cache. `from math import sqrt` does **not** load "only the `sqrt` part" — the entire `math` module is loaded either way.

**The only difference is which name gets bound in your namespace:**

* `import math` → binds one name, `math`, pointing at the **module object**. Everything inside is reached through the dot: `math.sqrt`, `math.pi`. The name `sqrt` by itself does not exist.
* `from math import sqrt` → binds the name `sqrt`, pointing **directly at the function**. The name `math` does not exist in your namespace, even though the module is fully loaded in memory.

So `import` gives you a **door with a label on it**; `from ... import` **takes specific items out of the room** and leaves the door unlabelled.

### ⚖️ Trade-offs (why this matters in practice)

| | `import math` | `from math import sqrt` |
| :--- | :--- | :--- |
| **Readability** | `math.sqrt(x)` says where the function came from | `sqrt(x)` is shorter, but the origin is invisible |
| **Namespace safety** | Safe — members stay behind the module name | Risky — can silently **overwrite** an existing name |
| **Typing effort** | Longer, repeated prefix | Concise for names used many times |
| **Loading cost** | Identical | Identical (no speed advantage) |

### ✅ Practical Guidance
* **Default to `import module`.** It is explicit and collision-proof.
* Use **`from ... import`** for a handful of names used very frequently, or when the member is buried in a long path — e.g. `from datetime import datetime` (the `datetime` *class* lives inside the `datetime` *module*, so `datetime.datetime.now()` becomes tiresome).
* Use **aliases** for community conventions: `import numpy as np`, `import pandas as pd`.
* **Never use `from module import *`** outside a throwaway REPL session. It dumps unknown names into your namespace and makes collisions impossible to trace.

### ⚙️ What Happens During `import`?
1. Python checks the `sys.modules` cache to see if the module is already loaded — if so, it is **not** re-executed (importing twice costs nothing).
2. If not cached, it searches the directories listed in `sys.path` (current directory first, then standard library, then `site-packages`).
3. It compiles the module into bytecode (`__pycache__/*.pyc`) and executes the module's top-level code once.
4. Finally, it binds the requested name(s) in your namespace — this last step is where the two forms differ.


````python
# Demonstrating Standard Library Imports
import math
from datetime import datetime, timezone
import sys

# 1. Math module functions
print(f"Pi Constant  : {math.pi}")
print(f"Factorial(6) : {math.factorial(6)}")
print(f"log10(1000)  : {math.log10(1000)}")

# 2. Datetime module
current_time = datetime.now(timezone.utc)
print(f"UTC Timestamp: {current_time.strftime('%Y-%m-%d %H:%M:%S %Z')}")

# 3. Module search path (sys.path)
print(f"\nPython looks for modules in {len(sys.path)} locations.")
print("The first three searched are:")
print(f" 1. {sys.path[0]}")
print(f" 2. {sys.path[1]}")
print(f" 3. {sys.path[2]}")
````

### 🧪 Seeing the Difference Concretely
The next cell proves three things: both forms give you the **same object**, the module is **fully loaded** either way, and `from ... import` can **silently shadow** a name you already had.


````python
import sys
import math
from math import sqrt

# 1. Same object, two different names
print(f"math.sqrt(64) = {math.sqrt(64)}   |   sqrt(64) = {sqrt(64)}")
print(f"Are they the same object? {sqrt is math.sqrt}")

# 2. 'from X import y' still loads ALL of X into memory
from fractions import Fraction
print(f"\nIs 'fractions' fully loaded in sys.modules? {'fractions' in sys.modules}")
print(f"Is the name 'fractions' usable here?          {'fractions' in globals()}")
print("  -> The whole module is in memory; we simply never named it.")

# 3. The shadowing danger of 'from X import y'
print(f"\nBuilt-in pow(2, 10)     = {pow(2, 10)}  (integer, supports 3 arguments)")

from math import pow          # this OVERWRITES the built-in pow!
print(f"After 'from math import pow':")
print(f"  pow(2, 10)            = {pow(2, 10)}  (now a float - a different function!)")

del pow                       # remove our shadow to restore the built-in
print(f"After 'del pow', built-in is back: pow(2, 10) = {pow(2, 10)}")
print("  -> With 'import math' this collision could never have happened.")
````

### 🛠️ Creating and Importing Your Own Custom Module
A module does **not** have to contain functions — it is simply a `.py` file whose names we want to reuse. One of the most common real-world uses is a **configuration module**: a single file holding the constants, thresholds and settings that every script in a project must agree on.

Let's build `icfoss_config.py` for a sensor deployment, then import it into this notebook session.

> ℹ️ *Functions inside modules are covered in the next module of this course. Everything below uses only constants, variables and expressions from Sections 1–4.*


````python
%%writefile icfoss_config.py
# icfoss_config.py - Shared Project Configuration Module
"""ICFOSS Urban Sensor Grid - Shared Constants and Settings"""
from typing import Final

# --- Project identity ---
PROJECT_NAME: Final[str] = "ICFOSS Urban Air Quality Grid"
DEPLOYMENT_CITY: Final[str] = "Thiruvananthapuram"
CONFIG_VERSION: Final[str] = "1.2"

# --- Physical constants ---
SEA_LEVEL_PRESSURE: Final[float] = 1013.25   # hPa
KELVIN_OFFSET: Final[float] = 273.15

# --- Operational thresholds ---
TEMP_WARNING_C: Final[float] = 35.0
TEMP_CRITICAL_C: Final[float] = 42.0
VOLTAGE_MIN_V: Final[float] = 3.0

# --- Derived settings (these expressions are evaluated at import time) ---
SAMPLING_HZ: Final[int] = 50
SAMPLING_INTERVAL_MS: Final[float] = 1000 / SAMPLING_HZ
NYQUIST_LIMIT_HZ: Final[float] = SAMPLING_HZ / 2

# Top-level code runs ONCE, the first time this module is imported.
print(f"[icfoss_config] Configuration v{CONFIG_VERSION} loaded.")
````

Now we import our custom module exactly like any standard library module. Watch for the `[icfoss_config]` message — it proves the module's top-level code was executed:


````python
# Importing and using our custom configuration module
import icfoss_config as cfg

# A sensor reading arriving from the field
reading_c = 36.4
reading_v = 3.12

print(f"\nProject        : {cfg.PROJECT_NAME}")
print(f"Location       : {cfg.DEPLOYMENT_CITY}")
print(f"Sampling Rate  : {cfg.SAMPLING_HZ} Hz (one sample every {cfg.SAMPLING_INTERVAL_MS} ms)")
print(f"Nyquist Limit  : {cfg.NYQUIST_LIMIT_HZ} Hz")

# The constants are used inside ordinary expressions
print(f"\nReading        : {reading_c} C = {reading_c + cfg.KELVIN_OFFSET:.2f} K")
print(f"Above warning threshold ({cfg.TEMP_WARNING_C} C)? {reading_c >= cfg.TEMP_WARNING_C}")
print(f"Voltage below minimum ({cfg.VOLTAGE_MIN_V} V)?    {reading_v < cfg.VOLTAGE_MIN_V}")
print(f"Module docstring: {cfg.__doc__}")

# Importing a SECOND time costs nothing - note that no [icfoss_config] message appears,
# because the module is served from the sys.modules cache (Section 6).
import icfoss_config
print("\nSecond import produced no load message -> served from cache.")
````

## 7. Integrated Classroom Demo: Single Sensor Node Health Report

A field node has just transmitted **one telemetry packet**. As with any network packet, every field arrives as **text** and must be converted before it can be used. Let's turn that raw packet into a readable engineering report, combining everything taught so far:

1. **Constants (Section 1):** `ALL_CAPS` thresholds declared with `typing.Final`.
2. **Data Types & Casting (Section 2):** raw string fields → `float` and `int`.
3. **Expressions (Section 3):** temperature conversions and a barometric altitude estimate.
4. **Operators (Section 4):** `**` for the altitude formula, `//` and `%` to break uptime into hours/minutes/seconds, comparison and logical operators for the health verdict.
5. **Imports (Section 6):** the standard library `datetime` module for a UTC timestamp.
6. **Formatted Output:** an aligned, professional report using f-string field widths.

> ℹ️ *Everything here is a single straight-line calculation. Loops, dictionaries and functions arrive in the next module of this course — they are what let you run this same logic over hundreds of packets at once.*


````python
# Comprehensive Integrated In-Class Demo - One Telemetry Packet, Start to Finish
from datetime import datetime, timezone
from typing import Final

# ---------------------------------------------------------------
# 1. CONSTANTS - thresholds and reference values
# ---------------------------------------------------------------
PROJECT_NAME: Final[str] = "ICFOSS Urban Air Quality Grid"
SEA_LEVEL_PRESSURE: Final[float] = 1013.25   # hPa
KELVIN_OFFSET: Final[float] = 273.15

TEMP_WARNING_C: Final[float] = 35.0
TEMP_CRITICAL_C: Final[float] = 42.0
VOLTAGE_MIN_V: Final[float] = 3.0

BAROMETRIC_SCALE: Final[float] = 44330.0
BAROMETRIC_EXPONENT: Final[float] = 1 / 5.255

# ---------------------------------------------------------------
# 2. THE RAW PACKET - every field arrives as TEXT
# ---------------------------------------------------------------
raw_node_id     = "NODE-BETA"
raw_temperature = "36.20"    # degrees Celsius
raw_voltage     = "3.15"     # volts
raw_pressure    = "1009"     # hectopascals
raw_uptime_secs = "9475"     # seconds since last reboot

# ---------------------------------------------------------------
# 3. CASTING - text into usable numbers
# ---------------------------------------------------------------
temperature_c = float(raw_temperature)
voltage_v     = float(raw_voltage)
pressure_hpa  = int(raw_pressure)
uptime_secs   = int(raw_uptime_secs)

# ---------------------------------------------------------------
# 4. EXPRESSIONS - derived engineering values
# ---------------------------------------------------------------
temperature_f  = temperature_c * 9 / 5 + 32
temperature_k  = temperature_c + KELVIN_OFFSET
pressure_delta = pressure_hpa - SEA_LEVEL_PRESSURE
altitude_m     = BAROMETRIC_SCALE * (1 - (pressure_hpa / SEA_LEVEL_PRESSURE) ** BAROMETRIC_EXPONENT)

# Floor division and modulo split a total into units (Section 4.1)
uptime_hours   = uptime_secs // 3600
uptime_minutes = (uptime_secs % 3600) // 60
uptime_seconds = uptime_secs % 60

# ---------------------------------------------------------------
# 5. LOGIC - comparison and logical operators decide the verdict
# ---------------------------------------------------------------
is_temp_warning  = temperature_c >= TEMP_WARNING_C
is_temp_critical = temperature_c >= TEMP_CRITICAL_C
is_power_fault   = voltage_v < VOLTAGE_MIN_V

if is_temp_critical or is_power_fault:
    status = "\U0001f6a8 CRITICAL"
elif is_temp_warning:
    status = "\u26a0\ufe0f  WARNING"
else:
    status = "\u2705 NORMAL"

# ---------------------------------------------------------------
# 6. FORMATTED REPORT
# ---------------------------------------------------------------
stamp = datetime.now(timezone.utc).strftime("%Y-%m-%d %H:%M:%S %Z")

print("=" * 56)
print(f"  {PROJECT_NAME}")
print(f"  Node Health Report - {raw_node_id}")
print("=" * 56)
print(f"  Generated (UTC)   : {stamp}")
print(f"  Uptime            : {uptime_hours}h {uptime_minutes}m {uptime_seconds}s")
print("-" * 56)
print(f"  Temperature       : {temperature_c:>10.2f} C")
print(f"                      {temperature_f:>10.2f} F")
print(f"                      {temperature_k:>10.2f} K")
print(f"  Supply Voltage    : {voltage_v:>10.2f} V")
print(f"  Pressure          : {pressure_hpa:>10} hPa")
print(f"  Deviation from SL : {pressure_delta:>10.2f} hPa")
print(f"  Estimated Altitude: {altitude_m:>10.1f} m")
print("-" * 56)
print(f"  Above warning ({TEMP_WARNING_C} C)?  {is_temp_warning}")
print(f"  Above critical ({TEMP_CRITICAL_C} C)? {is_temp_critical}")
print(f"  Power fault (< {VOLTAGE_MIN_V} V)?    {is_power_fault}")
print("=" * 56)
print(f"  OVERALL STATUS    : {status}")
print("=" * 56)

# CLASSROOM EXPERIMENT: change raw_voltage to "2.90" or raw_temperature to "43.5"
# and re-run. Which branch fires, and why?
````

## 🎓 Summary Checklist for Lectures 1 & 2:

1. **Variables are Pointers:** In Python, variables are names bound to objects in memory.
2. **Constants:** PEP 8 convention uses `ALL_CAPS` and `typing.Final` for immutable semantics.
3. **Type System:** Dynamic but strongly typed. Always cast strings to numbers (`int()`/`float()`) when taking user input.
4. **Operators:** Division `/` always returns float; integer division `//` rounds towards $-\infty$; modulo `%` preserves mathematical identities.
5. **Command Line:** Standalone scripts read parameters from `sys.argv` (always strings) — check `len(sys.argv)` before indexing it.
6. **Imports:** Modules isolate functionality into distinct namespaces. Always prefer explicit imports over wildcard imports (`*`).
