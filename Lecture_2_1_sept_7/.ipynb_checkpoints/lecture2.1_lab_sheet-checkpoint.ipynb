## Setup — run this first

This defines `verify()`, which checks your answers. It never stops the notebook, so
**Restart and Run All** still works with exercises unfinished.

````python
def verify(name, expected, tol=1e-9):
    """Check a student answer stored in a variable called `name`."""
    if name not in globals():
        print(f"[ ] {name} — not attempted yet")
        return
    got = globals()[name]
    if got is Ellipsis:
        print(f"[ ] {name} — not attempted yet")
        return
    numeric = (int, float, complex)
    if isinstance(got, numeric) and isinstance(expected, numeric) \
            and not isinstance(got, bool):
        ok = abs(got - expected) <= tol   # abs() of a complex gives its magnitude
    else:
        ok = got == expected
    if ok:
        print(f"[PASS] {name} = {got!r}")
    else:
        print(f"[FAIL] {name} = {got!r}   (expected {expected!r})")


def verify_type(name, expected_type):
    """Check that a variable exists and has the expected type."""
    if name not in globals() or globals()[name] is Ellipsis:
        print(f"[ ] {name} — not attempted yet")
        return
    actual = type(globals()[name])
    mark = "PASS" if actual is expected_type else "FAIL"
    print(f"[{mark}] {name} is {actual.__name__}, expected {expected_type.__name__}")


print("verify() and verify_type() are ready.")
````

# 2 · Python as a calculator

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | addition | `7 + 2` | `9` |
| `-` | subtraction | `7 - 2` | `5` |
| `*` | multiplication | `7 * 2` | `14` |
| `/` | division — **always gives a float** | `7 / 2` | `3.5` |
| `//` | floor division — discards the remainder | `7 // 2` | `3` |
| `%` | modulus — keeps only the remainder | `7 % 2` | `1` |
| `**` | exponentiation | `7 ** 2` | `49` |

Note the two that are easy to confuse: `/` and `//`. And note that `*` means multiply —
`x` is not an operator in Python.

````python
print(7 + 2)
print(7 - 2)
print(7 * 2)
print(7 / 2)      # note the .0 — division always produces a float
print(7 // 2)     # floor division: the whole part only
print(7 % 2)      # modulus: the remainder only
print(7 ** 2)
````

### Anything after a `#` is a comment

Python ignores it completely. A good comment says *why*, not *what*. `# add 1 to i` is
noise. `# skip the header row` is useful.

````python
# This whole line is ignored.
seconds_per_day = 24 * 60 * 60   # a trailing comment works too
print(seconds_per_day)
````

### Precedence

`**` first, then `*` `/` `//` `%`, then `+` `-`. Brackets override everything. Predict
both answers before running.

````python
print(2 + 3 * 4)
print((2 + 3) * 4)
````

# 3 · Variables

A variable is a name attached to a value. The `=` sign is **assignment**, not equality —
read `mass = 4.5` as "let the name *mass* refer to 4.5".

````python
mass = 4.5          # kilograms
acceleration = 9.81 # m/s^2

force = mass * acceleration
print(force)
````

### Naming

Names must start with a letter or underscore, and may contain letters, digits and
underscores. They are **case-sensitive** — `Force` and `force` are different names.

The convention is `lower_case_with_underscores`. Use it from today; marks in this course
include readability.

| Poor | Better | Why |
|---|---|---|
| `x` | `flow_rate` | says what it holds |
| `t1`, `t2` | `start_time`, `end_time` | says which is which |
| `TEMP` | `temperature_c` | includes the unit |
| `list` | `readings` | `list` is a built-in name — do not shadow it |

````python
# Several assignments at once
length, width, height = 2.5, 1.8, 0.4

volume = length * width * height
print("Volume:", volume, "cubic metres")

# A name can be reassigned — the old value is simply forgotten
volume = volume * 1000
print("Volume:", volume, "litres")
````

# 4 · Reading errors properly

You have now met five error types by accident. This section makes the skill deliberate.

### How to read a traceback

Read it **from the bottom up**:

1. **the last line** names the error type and gives a message — start here
2. **just above it**, Python shows the offending line with a marker under the problem
3. **above that**, the chain of calls that got there — ignore this for now

### The errors you will meet this month

| Error | Usually means | First thing to check |
|---|---|---|
| `SyntaxError` | Python could not translate the code at all | missing bracket, quote or colon |
| `IndentationError` | the spacing at the start of a line is wrong | stray or missing spaces |
| `NameError` | a name that does not exist yet | spelling; did you run the cell above? |
| `TypeError` | the operation does not work on that type | are you mixing text and numbers? |
| `ValueError` | right type, impossible value | is the text really a number? |
| `ZeroDivisionError` | division by zero | what made the divisor zero? |
| `IndexError` | a position that is not there | counting starts at 0 |
| `KeyError` | a dictionary key that is not there | check the spelling of the key |
| `ModuleNotFoundError` | the import failed | spelling; correct environment? |

The one skill worth building today: **read the error type first, before looking at your
code.** The type tells you which of the nine situations above you are in, and that narrows
the search enormously.

# 5 · The built-in data types

You do not need to master these today. The goal is recognition.

| Type | Written as | Holds |
|---|---|---|
| `int` | `42`, `-7` | whole numbers |
| `float` | `3.14`, `9.81` | decimal numbers |
| `complex` | `2 + 3j` | complex numbers, with `j` for the imaginary unit |
| `bool` | `True`, `False` | one of exactly two values — note the capitals |
| `str` | `'text'`, `"text"` | a sequence of characters |
| `list` | `[1, 2, 3]` | an ordered, changeable collection |
| `tuple` | `(1, 2, 3)` | an ordered collection that cannot be changed |
| `set` | `{1, 2, 3}` | unordered, no duplicates |
| `dict` | `{'k': 'v'}` | key–value pairs |

````python
sample_count   = 42                                  # int
gravity        = 9.81                                # float
impedance      = 3 + 4j                              # complex
is_calibrated  = True                                # bool
station_name   = "Thiruvananthapuram"                # str
readings       = [21.4, 22.1, 20.9, 23.6]            # list
coordinates    = (8.5241, 76.9366)                   # tuple
unique_ids     = {"A1", "B2", "A1", "C3"}            # set — note the duplicate
station_record = {"name": "TVM", "readings": 4}      # dict

print(unique_ids)      # the repeated "A1" appears only once
print(readings[0])     # counting starts at 0, not 1
````

### Checking a type with `type()`

````python
for value in (sample_count, gravity, impedance, is_calibrated,
              station_name, readings, coordinates, unique_ids, station_record):
    print(type(value))
````

### One result worth staring at

````python
print(type(10))        # int
print(type(10.0))      # float — the .0 changes the type
print(type("10"))      # str  — quotes make it text, not a number

print(10 + 10.0)       # fine: Python promotes the int to a float
````

### Converting between types

`int()`, `float()`, `str()` and `bool()` build a **new** value of that type. They do not
change the original.

````python
print(int("42"), type(int("42")))       # text -> whole number
print(float("3.14"))                    # text -> decimal
print(str(99) + " bottles")             # number -> text, so + can join it
print(int(7.9))                         # truncates toward zero — it does NOT round
print(round(7.9))                       # this is what rounds
````

**`int(7.9)` gives 7, not 8.** It discards the fractional part rather than rounding. This
causes real bugs. When you want rounding, say `round()`.

### The debugging habit to take away

When something fails, in this order:

1. read the **error type** on the last line — it tells you which kind of problem you have
2. find the **line number** and look at that line, not the whole cell
3. ask what the code *actually* says, not what you meant it to say
4. only then start changing things — and change one thing at a time

An unsolved error you have thought about is worth more than a solved one you copied.
