In standard sequential execution, instructions execute strictly from top to bottom. **Decision statements** alter this flow by evaluating Boolean conditions and selecting which execution branch to take.

### Python Branching Structures:
1. **One-Way Decision (`if`)**: Executes a block if and only if the condition evaluates to `True`.
2. **Two-Way Decision (`if-else`)**: Executes one block if `True`, and an alternate block if `False`.
3. **Multi-Way Decision (`if-elif-else`)**: Evaluates a chain of conditions in top-down order; executes the first true branch and skips the rest.
4. **Conditional Expression (Inline Ternary)**: `result = true_value if condition else false_value`.

# Part 2: The `while` Loop — Condition-Controlled Iteration


A `while` loop repeatedly executes a block of statements as long as its Boolean condition evaluates to `True`.

 
! 
### The Three Pillars of a Safe `while` Loop:
1. **Initialization**: Loop control variable must be set before the loop starts.
2. **Condition**: Must eventually become `False`.
3. **Update Step**: Control variable must be updated in the loop body. Missing this causes an **infinite loop**!


````python
# 2.1 Basic while Loop: Counter Progression
count = 1  # 1. Initialization

print("Starting countdown...")
while count <= 5:  # 2. Test condition
    print(f"  Step {count} of 5")
    count += 1     # 3. Update step (Increments count)

print("Loop finished! Final count value:", count)
````

````python
# 2.2 Mathematical Digit Extraction Loop Pattern
# Very important pattern: Processing integers digit-by-digit from right to left!
# Given n = 9845:
#   n % 10  extracts the last digit (5)
#   n // 10 strips the last digit (leaving 984)

number = 9845
temp = number
digit_sum = 0
digit_count = 0

print(f"Decomposing number: {number}")
while temp > 0:
    last_digit = temp % 10       # Extract rightmost digit
    digit_sum += last_digit      # Accumulate sum
    digit_count += 1             # Count digit
    print(f"  Extracted: {last_digit} | Remaining: {temp // 10}")
    temp = temp // 10            # Discard rightmost digit (Update step)

print(f"Total digits: {digit_count} | Sum of digits: {digit_sum}")
````

````python
# 2.3 Sentinel-Controlled while Loop
# A sentinel value signals the termination of data entry
# Here we simulate reading student test scores until -1 is entered:

simulated_scores = [78, 85, 92, 64, 88, -1]  # -1 is the sentinel!
idx = 0
total_score = 0
valid_entries = 0

score = simulated_scores[idx]
idx += 1

while score != -1:
    total_score += score
    valid_entries += 1
    score = simulated_scores[idx]  # Read next score
    idx += 1

average = total_score / valid_entries if valid_entries > 0 else 0
print(f"Processed {valid_entries} scores | Total: {total_score} | Average: {average:.2f}")
````

# Part 3: The `for` Loop & The `range()` Generator


In Python, the `for` loop is an **iterator-based loop**. It iterates directly across the elements of any sequence (strings, lists, tuples, or numeric ranges).

### 1. Syntax of `for` Loop:
```python
for item in sequence:
    # Execute code using item
```

### 2. The `range([start], stop, [step])` Built-in:
Generates an arithmetic progression of integers on demand (memory efficient!):
- `start` (Optional, default = $0$): First number in sequence (inclusive).
- `stop` (Mandatory): Upper bound (exclusive — stops at $\text{stop} - 1$).
- `step` (Optional, default = $1$): Increment between successive numbers (can be negative for countdowns!).

| `range()` Invocation | Sequence Generated |
| :--- | :--- |
| `range(5)` | $0, 1, 2, 3, 4$ |
| `range(2, 7)` | $2, 3, 4, 5, 6$ |
| `range(1, 10, 2)` | $1, 3, 5, 7, 9$ (Odd numbers) |
| `range(10, 0, -2)` | $10, 8, 6, 4, 2$ (Descending even numbers) |


````python
# 3.1 Exploring range() Parameters
print("1. range(5)             :", list(range(5)))
print("2. range(3, 8)          :", list(range(3, 8)))
print("3. range(10, 50, 10)    :", list(range(10, 50, 10)))
print("4. range(10, 0, -1)     :", list(range(10, 0, -1)))
print("5. range(-5, 5, 2)      :", list(range(-5, 5, 2)))
````

# Part 4: Loop Control Statements — `break`, `continue`, `pass` & `else`


Python provides statements to alter the natural execution cycle of loops.

### Summary of Control Statements:
| Statement | Action | Flowchart Meaning |
| :---: | :--- | :--- |
| `break` | **Terminates** loop immediately | Jumps completely out of loop to the statement following the loop |
| `continue` | **Skips** current iteration | Aborts remainder of loop body and jumps directly to next iteration |
| `pass` | **No-op placeholder** | Does nothing; syntactically required where a statement block is expected |
| `loop...else` | **Post-completion block** | Executes **ONLY if the loop completed naturally** without hitting `break`! |

```
      for item in seq:                    for item in seq:
          if condition:                       if condition:
              break  ==========+                  continue ----+
          # statements        |               # statements     |
                              |                                | (Next iteration)
          <-------------------+               <----------------+
```



! 



````python
# 4.1 The break Statement: Early Exit
# Scenario: Search for the first negative sensor reading and stop immediately

sensor_readings = [24.5, 23.8, 25.1, -99.0, 26.4, 28.0]

for idx, reading in enumerate(sensor_readings):
    if reading < 0:
        print(f"FAULT DETECTED at index {idx}: Invalid reading {reading}°C!")
        break  # Terminate loop immediately!
    print(f"  Reading {idx}: {reading}°C normal.")

print("Post-sensor audit complete.")
# To Do
# try changing one sensor _reading to -1 and rerun
````

````python
# 4.2 The continue Statement: Skipping Unwanted Iterations
# Scenario: Process only valid positive readings, skip negative noise/errors

sensor_readings = [24.5, -99.0, 25.1, -1.0, 26.4]
valid_sum = 0
valid_count = 0

for reading in sensor_readings:
    if reading < 0:
        print(f"  Skipping erroneous reading: {reading}")
        continue  # Skip to the next iteration; lines below are bypassed!

    valid_sum += reading
    valid_count += 1
    print(f"  Processed valid reading: {reading}")

print(f"Processed {valid_count} valid readings. Mean = {valid_sum / valid_count:.2f}°C")
# To Do
# try changing one sensor _reading to -1 and rerun
````

````python
# 4.3 The Unique Python loop...else Construct!
# In Python, for and while loops can have an 'else' block!
# The 'else' block executes ONLY if the loop finishes without encountering 'break'.
# Perfect for search operations!

target = 42
data_stream = [10, 25, 33, 78, 99, 14]

for item in data_stream:
    if item == target:
        print(f"Target {target} was found in the dataset!")
        break
else:
    # This executes ONLY if the loop ran to completion without finding target!
    print(f"Target {target} was NOT found in the dataset.")
````

# Part 5: Nested Loops & Geometric Patterns


When one loop is enclosed completely inside the body of another loop, it is termed a **nested loop**.

### Execution Dynamics:
- For **every single iteration** of the outer loop, the inner loop executes through its **entire iteration range**.
- If outer loop runs $M$ times and inner loop runs $N$ times, the innermost block executes $M \times N$ times!

```python
for i in range(1, 4):      # Outer Loop (i = 1, 2, 3)
    for j in range(1, 3):  # Inner Loop (j = 1, 2)
        print(f"i={i}, j={j}")
```


````python
# 5.1 Nested Loops: 2D Grid Coordinates and Multiplication Matrix
# Printing a neat 5x5 multiplication table:

print("     ", end="")
for col in range(1, 6):
    print(f"{col:>5}", end="")
print("\n" + "    " + "-" * 26)

for row in range(1, 6):
    print(f"{row:>3} |", end="")
    for col in range(1, 6):
        product = row * col
        print(f"{product:>5}", end="")
    print()  # Newline after inner loop finishes
````

````python
# 5.2 Geometric Pattern Generation
# Triangle of Stars:
# *
# * *
# * * *
# * * * *
# * * * * *

rows = 5
print("Right-Angled Triangle Pattern:")
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print("* ", end="")
    print()  # Move to next row

print("\nInverted Number Triangle:")
for i in range(rows, 0, -1):
    for j in range(1, i + 1):
        print(f"{j} ", end="")
    print()
````

# Part 6: Iterating Through Dictionaries
 Teaching note :Remind students about dictionary

Dictionaries provide three distinct iterable views:
1. `.keys()`: Iterates across keys (default behavior when iterating `for k in d`).
2. `.values()`: Iterates across values.
3. `.items()`: Iterates across `(key, value)` tuple pairs simultaneously (recommended!).

```python
dataset = {"Epoch_1": 0.85, "Epoch_2": 0.91, "Epoch_3": 0.94}

for epoch, accuracy in dataset.items():
    print(f"{epoch} -> Accuracy: {accuracy:.1%}")
```


````python
# 6.1 Dictionary Iteration Patterns
course_credits = {
    "Programming in Python": 3,
    "Applied Machine Learning": 4,
    "Linear Algebra for AI": 3,
    "Ethics in AI": 1
}

print("--- 1. Default / Keys View ---")
for course in course_credits:
    print(f"  Course: {course}")

print("\n--- 2. Values View ---")
total_credits = 0
for cred in course_credits.values():
    total_credits += cred
print(f"  Total Credits Required: {total_credits}")

print("\n--- 3. Items View (Key + Value Unpacking) ---")
for course, cred in course_credits.items():
    print(f"  {course:<26} : {cred} Credit(s)")
````

````python
# 6.2 Filtering and Transforming Dictionaries with Loops
# Scenario: Identify all courses carrying >= 3 credits
high_credit_courses = {}

for course, cred in course_credits.items():
    if cred >= 3:
        high_credit_courses[course] = cred

print("Core High-Credit Courses:")
for c, cr in high_credit_courses.items():
    print(f"  * {c} ({cr} credits)")

# Advanced: Same operation via Dictionary Comprehension!
high_credit_comp = {c: cr for c, cr in course_credits.items() if cr >= 3}
assert high_credit_courses == high_credit_comp
print("Dictionary comprehension matches iterative loop successfully.")
````
