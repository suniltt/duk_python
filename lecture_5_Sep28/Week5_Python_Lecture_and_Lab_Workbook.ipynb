### Why Use Functions to Avoid Code Repetition? (DRY Principle)

One of the most significant benefits of using functions is adhering to the **DRY (Don't Repeat Yourself)** principle. When you find yourself writing the same or very similar blocks of code multiple times, it's a strong indicator that a function is needed.

**Benefits of avoiding repetition with functions:**

1.  **Readability:** Your code becomes cleaner and easier to understand, as the logic for a specific task is encapsulated in one place.
2.  **Maintainability:** If you need to change or fix a bug in a particular piece of logic, you only have to do it in one place (the function definition), rather than searching and modifying every instance of that code throughout your program.
3.  **Reusability:** Functions can be called multiple times from different parts of your program, or even imported and used in other programs.
4.  **Reduced Errors:** By centralizing logic, you reduce the chances of introducing inconsistencies or errors when making changes.

#### Example: Calculating Area of Circles (Without vs. With Function)

Let's consider a simple task: calculating the area of different circles. Without a function, if you need to do this multiple times, you would repeat the calculation logic directly each time.

````python
import math

# Calculate area for radius 5
radius1 = 5
area1 = math.pi * (radius1 ** 2)
print(f"Area of circle with radius {radius1}: {area1:.2f}")

# Calculate area for radius 10
radius2 = 10
area2 = math.pi * (radius2 ** 2)
print(f"Area of circle with radius {radius2}: {area2:.2f}")

# Calculate area for radius 7.5
radius3 = 7.5
area3 = math.pi * (radius3 ** 2)
print(f"Area of circle with radius {radius3}: {area3:.2f}")

# Notice the repetition of 'math.pi * (radius ** 2)'
````

#### A Text Formatting Utility

Imagine you often need to format strings in a specific way: perhaps capitalizing the first letter, ensuring it doesn't exceed a certain length (and adding an ellipsis if it does), or stripping whitespace. Without a function, you'd repeat these steps every time.

This becomes cumbersome and error-prone very quickly. A function elegantly encapsulates this logic:

````python
# With a function

def format_text(text, max_length=10):
    """Strips whitespace, capitalizes, and truncates text with ellipsis if too long."""
    processed_text = str(text).strip().capitalize()
    if len(processed_text) > max_length:
        processed_text = processed_text[:max_length - 3] + "..."
    return processed_text

print(f"Formatted Text 1: {format_text('  hello world  ')}")
print(f"Formatted Text 2: {format_text('  this is a very long sentence that needs to be truncated  ', max_length=20)}")
print(f"Formatted Text 3: {format_text(1234567890123, max_length=15)}")

# The formatting logic is defined once and easily reused with different parameters.
````

# Part 2: Function Arguments — How to Give Functions Information

Think of a function as a mini-program that does a specific job. To do its job, it often needs some information, like ingredients for a recipe. These pieces of information you give to the function are called **arguments**.

Python offers a few flexible ways to hand over these "ingredients" to your functions:

### 1. Positional Arguments (Order Matters Like a Checklist!)

Imagine you have a checklist for ingredients: first flour, then sugar. With positional arguments, you just list the values in that exact order, and Python knows which value goes with which item on the function's internal checklist.

```python
def bake_cake(flour, sugar):
    print(f"Baking a cake with {flour} and {sugar}.")

bake_cake("2 cups flour", "1 cup sugar") # '2 cups flour' is first, so it goes to 'flour'. '1 cup sugar' is second, so it goes to 'sugar'.
```

### 2. Keyword (Named) Arguments (Calling Out Ingredients by Name!)

Instead of just listing ingredients, imagine you label each one as you hand it over: "Here's the flour!" "Here's the sugar!" With keyword arguments, you specify which internal placeholder (parameter) each value belongs to by using its name (e.g., `sugar="1 cup"`). The cool part? The order doesn't matter!

```python
def bake_cake(flour, sugar):
    print(f"Baking a cake with {flour} and {sugar}.")

bake_cake(sugar="1 cup sugar", flour="2 cups flour") # Python knows 'sugar' is 'sugar' and 'flour' is 'flour', no matter the order.
```

### 3. Default Parameters (Optional Ingredients!)

Sometimes, a function has parameters that often take the same value, or where providing a value is optional. For these, you can set "default" values. If you don't provide a value for them, the function uses its default.

```python
def greet(name, message="Hello"):
    print(f"{message}, {name}!")

# Here, 'name' is required, but 'message' has a default of "Hello".

greet("Alice")               # Uses default message: "Hello, Alice!"
greet("Bob", "Good morning")  # Overrides default: "Good morning, Bob!"
```

-   **Important Rule**: Any parameters with default values **must** come *after* parameters that don't have default values in your function's definition. This makes it clear which inputs are always needed.

### 4. Flexible Arguments (`*args` and `**kwargs`) (The "Anything Else?" Catch-alls!)

What if your main recipe is set, but you also want to allow *any number* of extra items or special instructions? These are your "catch-all" arguments:

-   `*args` (pronounced "star args"): This lets you pass any number of **extra items in order**. They get collected into a group (a `tuple`) inside your function.
    *   **Example**: Imagine ordering a pizza. You might say `order_pizza("large", "pepperoni", "mushrooms", "onions")`. Here, `"pepperoni"`, `"mushrooms"`, `"onions"` are extra toppings passed in order, collected by `*args`.

-   `**kwargs` (pronounced "star star kwargs"): This lets you pass any number of **extra named items**. They get collected into a list of labels and values (a `dictionary`) inside your function.
    *   **Example**: Continuing with the pizza, you might add `order_pizza("large", special_instruction="extra crispy crust", delivery_time="7 PM")`. Here, `special_instruction` and `delivery_time` are extra named instructions, collected by `**kwargs`.

These flexible arguments are super useful when you don't know exactly how many extra bits of information a function might need!

### The Scenario: Creating a User Profile

Imagine you're building a system where users can create profiles. This function will help collect and display user information.

````python
# This is our main function, like a recipe for creating a user's profile.
# It takes different pieces of information (ingredients) about a user.
# Some ingredients are required, some are optional, and some can be extra.
def create_user_profile(username, email, age=None, city='Unknown', *interests, **extra_info):
    """
    Creates a user profile with basic and optional information.

    Parameters:
        username (str): The user's unique username (always needed, in order).
        email (str): The user's email address (always needed, in order).
        age (int, optional): The user's age. If not given, it's 'None'.
        city (str, optional): The user's city. If not given, it's 'Unknown'.
        *interests (str): Any number of extra hobbies or interests (like extra toppings on a pizza).
        **extra_info (dict): Any other custom details (like 'country' or 'occupation').
    """
    # We start building the user's profile with the basic information.
    profile = {
        "Username": username,
        "Email": email,
        "Age": age,
        "City": city,
        "Interests": list(interests) # Collects all the extra interests into a list.
    }
    # If there are any other custom details (like country), add them to the profile.
    profile.update(extra_info)
    return profile # Send back the completed user profile.
````

#### 1. Positional Arguments (Order Matters)

These are the most common! You pass values in the exact order the function expects them. Python matches them from left to right.

````python
# 'Alice123' goes to `username`, 'alice@example.com' goes to `email`
profile1 = create_user_profile("Alice123", "alice@example.com")
print("Profile 1 (Positional):", profile1)

# We can also pass 'age' (30) and 'city' ('New York') positionally.
# They fill the next available spots in the function's expected order.
profile2 = create_user_profile("Bob_Z", "bob@email.com", 30, "New York")
print("Profile 2 (Positional with optional):", profile2)
````

#### 2. Default Values (Optional Inputs)

Some function parameters have a 'default' value. If you don't provide a value for them, the function uses the default. You can also *override* the default.

````python
# No age or city provided, so `age=None` and `city='Unknown'` are used automatically.
profile3 = create_user_profile("CharlieD", "charlie@web.com")
print("Profile 3 (Defaults):", profile3)

# Here, we 'override' the default city. Even though 'city' has a default, we give it 'London'.
profile4 = create_user_profile("DianaP", "diana@mail.net", city="London")
print("Profile 4 (Overriding Default):", profile4)
````

---

#### 3. Keyword (Named) Arguments (Order Does Not Matter)

With keyword arguments, you specify the parameter's name when you pass the value (e.g., `age=25`). This makes your code very clear, and you don't have to worry about the order (except for positional arguments, which must come first).

````python
# The order of `age` and `city` doesn't matter here because we named them
profile5 = create_user_profile("Eve_R", "eve@abc.org", city="Paris", age=28)
print("Profile 5 (Keyword):", profile5)

# You can mix both positional and keyword arguments, but positional ones MUST come first.
profile6 = create_user_profile("Frankie", "frankie@xyz.com", 40, city="Berlin")
print("Profile 6 (Mixed):", profile6)
````

#### 4. Flexible Arguments: `*args` (Many Positional) and `**kwargs` (Many Named)

Sometimes you want a function to accept *any number* of extra inputs, either in order or by name. Python has special ways to handle this:

*   `*args`: Collects any extra **positional** arguments into a **tuple**.
    *   Think of it as a "catch-all" for extra items that are just listed one after another.
*   `**kwargs`: Collects any extra **keyword** arguments into a **dictionary**.
    *   Think of it as a "catch-all" for extra items that are given a `name=value` format.

````python
# `*interests` (called `*args`): Gathers any extra unnamed items into a list of 'interests'.
# Notice 'age' (35) and 'city' ('Tokyo') are given in their expected positional order first.
# Then, 'coding', 'reading', 'travel' are extra hobbies that get collected by `*interests`.
profile7 = create_user_profile(
    "HenryK", "henry@dev.io",
    35, "Tokyo", # age and city passed positionally
    "coding", "reading", "travel" # These are extra items for `*interests`
)
print(f"Profile 7 (with *args):\n{profile7}")

# `**extra_info` (called `**kwargs`): Gathers any extra named (keyword) items into a dictionary.
# Here, 'country' and 'occupation' are custom, named details that go into `**extra_info`.
profile8 = create_user_profile(
    "IvyL", "ivy@corp.net",
    age=22,
    country="Canada", occupation="Student" # These are extra named items for `**extra_info`
)
print(f"Profile 8 (with **kwargs):\n{profile8}")

# Combining both `*args` and `**kwargs`.
# Again, 'age' and 'city' are given positionally first.
# Then, 'gaming' and 'streaming' go into `*interests`.
# Finally, 'membership_level' and 'last_login' go into `**extra_info`.
profile9 = create_user_profile(
    "JackW", "jack@gaming.net",
    25, "Seoul", # age and city passed positionally
    "gaming", "streaming", # These are extra items for `*interests`
    membership_level="Premium", last_login="2023-11-01" # These are extra named items for `**extra_info`
)
print(f"Profile 9 (with *args and **kwargs):\n{profile9}")
````

Understanding these ways to pass arguments gives you a lot of flexibility when designing and using functions in Python!

````python
# 2.4 Flexible / Arbitrary Arguments: *args and **kwargs

# 1. *args: Pack arbitrary positional arguments into a tuple
def compute_arithmetic_mean(*values):
    """Computes mean of any count of numbers passed as arguments."""
    print(f"  Received *args as type '{type(values).__name__}': {values}")
    if not values:
        return 0.0
    return sum(values) / len(values)

print("Mean of 3 numbers:", compute_arithmetic_mean(10, 20, 30))
print("Mean of 5 numbers:", compute_arithmetic_mean(5, 10, 15, 20, 25))

# 2. **kwargs: Pack arbitrary keyword arguments into a dictionary
def print_experiment_config(model_type, **hyperparameters):
    """Logs experiment configuration parameters dynamically."""
    print(f"\nModel Architecture: {model_type}")
    print("Hyperparameters:")
    for param, val in hyperparameters.items():
        print(f"  * {param:<15}: {val}")

print_experiment_config("Deep-ResNet", lr=0.001, batch_size=64, optimizer="Adam", epochs=50)
````

### Introduction to Variable Scope in Python

When you create a variable in Python, where you create it determines where in your program that variable can be seen and used. This concept is called **scope**.

Think of it like this:

*   **Global Scope**: Variables defined directly in the main part of your script (outside any functions) are like announcements made over a loudspeaker throughout an entire building. Everyone in the building can hear and understand them.
*   **Local Scope**: Variables defined inside a function are like notes written inside a specific room. Only people inside that room can see and use those notes. Once you leave the room, the notes are no longer visible.

Understanding scope is crucial for writing functions that interact correctly with data in your program without causing unexpected side effects. We'll explore this more in the next section!

# Part 3: Variable Scope & The LEGB Rule

**Scope** defines the region of a program where a particular variable name is visible and accessible.

```
       +-----------------------------------------------------------+
       | Built-in (B): Python pre-loaded names (len, print, int)   |
       |   +---------------------------------------------------+   |
       |   | Global (G): Top-level module / notebook variables |   |
       |   |   +-------------------------------------------+   |   |
       |   |   | Enclosing (E): Outer enclosing functions  |   |   |
       |   |   |   +-----------------------------------+   |   |   |
       |   |   |   | Local (L): Inside current function|   |   |   |
       |   |   |   +-----------------------------------+   |   |   |
       |   |   +-------------------------------------------+   |   |
       |   +---------------------------------------------------+   |
       +-----------------------------------------------------------+
               Python searches outward: L -> E -> G -> B
```

### Key Scoping Rules:
1. **Local Scope**: Variables assigned inside a function exist only while that function is executing.
2. **Reading Global Variables**: Functions can freely *read* global variables without declaring them.
3. **Modifying Global Variables**: If a function assigns to a variable name, Python treats it as *local* by default. To modify a global variable, you must explicitly declare it with `global variable_name`.
4. **Variable Shadowing**: When a local variable shares the same name as a global variable, the local variable hides (shadows) the global one within that function's scope.


````python
# 3.1 Local vs Global: Shadowing and Isolation
system_mode = "PRODUCTION"  # Global variable

def check_environment():
    system_mode = "TESTING"  # Local variable! Shadows the global variable
    print("Inside function (local scope)  : system_mode =", system_mode)

check_environment()
print("Outside function (global scope): system_mode =", system_mode)
# Notice: The global variable was NOT modified!
````

````python
# 3.2 The 'global' Keyword: Modifying Module-Level State
transaction_counter = 0  # Global state

def process_transaction(amount):
    global transaction_counter  # Declare intent to modify the global variable!
    transaction_counter += 1
    print(f"Transaction #{transaction_counter} processed: Rs. {amount:,.2f}")

process_transaction(5000)
process_transaction(12000)
process_transaction(3500)
print(f"Total Transactions Logged in Global State: {transaction_counter}")

# Software Engineering Best Practice Note:
# In production software, minimize global mutable state! Prefer returning values instead.
````

# Part 4: Nested (Inner) Functions & The `nonlocal` Keyword


In Python, you can define functions inside other functions, pass them as arguments, and return them from functions.

```python
def outer_function(x):
    def inner_helper(y):
        return y * 2
    return inner_helper(x) + 5
```

### Why Use Nested Functions?
1. **Encapsulation & Helper Functions**: Hide algorithmic helper sub-routines from the outside world when they are only relevant inside the parent function.
2. **State Preservation (Closures)**: An inner function can remember values from its enclosing scope even after the outer function has finished executing!
3. **The `nonlocal` Keyword**: Used inside an inner function to rebind/modify a variable that belongs to the *enclosing (outer)* scope.


````python
# 4.1 Nested Helper Functions for Clean Decomposition
def process_student_transcript(name, theory_mark, lab_mark):
    """Computes student result with encapsulated grade determination."""

    # Nested inner function (invisible outside of process_student_transcript):
    def assign_letter_grade(percentage):
        if percentage >= 90: return "A+"
        elif percentage >= 80: return "A"
        elif percentage >= 70: return "B+"
        elif percentage >= 60: return "B"
        elif percentage >= 50: return "C"
        else: return "F"

    total = theory_mark + lab_mark
    pct = (total / 200) * 100
    grade = assign_letter_grade(pct)  # Call the nested helper

    return {
        "name": name,
        "total": total,
        "percentage": round(pct, 2),
        "grade": grade
    }

student_card = process_student_transcript("Bhavana", 88, 94)
print("Processed Card:", student_card)

# Note: Calling assign_letter_grade directly from outside raises a NameError!
# assign_letter_grade(85)  # -> NameError: name 'assign_letter_grade' is not defined
````

# Part 5: The `return` Statement & Multi-Value Unpacking


The `return` statement immediately terminates function execution and transfers control (along with an optional payload) back to the caller.

### Multi-Value Returns in Python:
Unlike C or Java (which restrict functions to returning a single primitive or reference), Python allows returning multiple comma-separated values:
```python
return min_val, max_val, avg_val
```
Under the hood, Python **packs** these values into a single `tuple`. The caller can unpack them immediately into distinct variables:
```python
low, high, mean = analyze_data(readings)
```


````python
# 5.1 Returning Multiple Values via Tuple Packing & Unpacking
def compute_vector_statistics(data_points):
    """Computes count, sum, mean, and span (max - min) of a dataset."""
    if not data_points:
        return 0, 0.0, 0.0, 0.0  # Guard against empty list

    n = len(data_points)
    total = sum(data_points)
    avg = total / n
    span = max(data_points) - min(data_points)

    # Comma-separated return automatically bundles into a tuple!
    return n, total, avg, span

sample_readings = [14.2, 18.5, 12.0, 19.8, 15.5]

# Direct multi-variable unpacking:
count, total, mean_val, spread = compute_vector_statistics(sample_readings)

print("Dataset:", sample_readings)
print(f"Count   : {count}")
print(f"Total   : {total:.2f}")
print(f"Mean    : {mean_val:.2f}")
print(f"Spread  : {spread:.2f}")
````

# Part 6: Recursive Functions & The Call Stack


A **Recursive Function** is a function that solves a problem by calling itself with smaller sub-instances of the same problem.

### The Two Mandatory Components of Every Recursive Function:
1. **Base Case (Anchor)**: The simplest, trivial condition that returns a direct result **without making further recursive calls**.
2. **Recursive Step**: The case where the function calls itself with an argument strictly closer to the base case.

```
                  factorial(3)
                   |  3 * factorial(2)
                   |       |  2 * factorial(1)
                   |       |       |  1 * factorial(0)
                   |       |       |       |
                   |       |       |       v Base Case returns 1
                   |       |       +---> 1 * 1 = 1
                   |       +-----------> 2 * 1 = 2
                   +-------------------> 3 * 2 = 6
```


````python
# 6.1 Classic Recursion: Factorial
# Mathematical definition:
#   0! = 1                  (Base case)
#   n! = n * (n - 1)!       (Recursive step)

def factorial_recursive(n):
    """Computes n! recursively with call-stack tracing."""
    # Base Case:
    if n == 0 or n == 1:
        return 1

    # Recursive Case:
    return n * factorial_recursive(n - 1)

for num in [0, 1, 5, 7]:
    print(f"{num}! = {factorial_recursive(num)}")
````

````python
# 6.2 Fibonacci Numbers: Recursive Implementation vs Performance Caveat
# F(0) = 0, F(1) = 1, F(n) = F(n-1) + F(n-2)

def fibonacci_recursive(n):
    """Computes nth Fibonacci term recursively."""
    # Base cases:
    if n == 0:
        return 0
    if n == 1:
        return 1

    # Recursive step:
    return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)

print("First 8 Fibonacci terms (Recursive):")
for i in range(8):
    print(f"F({i}) = {fibonacci_recursive(i)}")

# PERFORMANCE NOTE:
# Naive recursion for Fibonacci has O(2^n) exponential time complexity due to
# redundant sub-calls! In production/lab, iterative loops or memoization are preferred.
````
