## Problem 1: Core Data Types, Operator Precedence & Formatted Output (5 Marks)

### Problem Description & Scenario
An electrical energy monitoring unit at an EV charging station records the peak voltage ($V_0$) and peak current ($I_0$) as raw text strings from a serial sensor, along with the charging duration in hours or seconds. You need to compute the average Alternating Current (AC) power, total energy consumed in kilowatt-hours ($	ext{kWh}$), and generate a formatted billing receipt string.

### Part A: Syntax & Logical Flow Correction (2 Marks)
The function `debug_power_calc(voltage_str, current_str, hours)` is intended to:
1. Convert the input strings `voltage_str` (e.g., `"230.0"`) and `current_str` (e.g., `"4.0"`) into floating-point numbers.
2. Calculate the average AC power using the RMS formula:
   $$P_{	ext{avg}} = rac{V_0 	imes I_0}{\sqrt{2} 	imes \sqrt{2}} = rac{V_0 	imes I_0}{2}$$
3. Calculate the energy consumed in kilowatt-hours:
   $$E_{	ext{kWh}} = rac{P_{	ext{avg}} 	imes 	ext{hours}}{1000}$$
4. Check whether `hours` is equal to `0` (`is_zero`) and format a summary string `f"Energy: {energy_kwh:.2f} kWh | Zero-Hours: {is_zero}"`.

**Defects to Fix in Part A**:
- **Bug 1 (Type Error)**: `voltage_str` and `current_str` are multiplied directly as strings without converting them to `float`.
- **Bug 2 (Operator Precedence Error)**: The denominator is written as `v * i / (2 ** 0.5) * (2 ** 0.5)`. Because `/` and `*` have equal precedence and evaluate left-to-right, Python divides by $\sqrt{2}$ and then multiplies the numerator by $\sqrt{2}$ instead of dividing by the product $(\sqrt{2} 	imes \sqrt{2})$.
- **Bug 3 (Syntax Error)**: Inside the boolean check `is_zero = (hours = 0)`, the single assignment operator `=` is used instead of the equality comparison operator `==`.

### Part B: Skeleton Implementation (3 Marks)
Complete the function `decompose_duration_and_cost(total_seconds, rate_per_hour)`:
- **Input Parameters**:
  - `total_seconds` (`int`): Total charging time in seconds (e.g., `8130`).
  - `rate_per_hour` (`float`): Tariff rate in Rupees per hour (e.g., `50.0`).
- **Required Operations**:
  1. Using **only integer floor division (`//`) and modulo (`%`)**, break down `total_seconds` into whole `hours` ($1	ext{ hour} = 3600	ext{ seconds}$), whole `minutes` ($1	ext{ minute} = 60	ext{ seconds}$), and `seconds` (remaining seconds).
  2. Calculate the exact proportional cost for the session:
     $$	ext{total\_cost} = \left(rac{	ext{total\_seconds}}{3600.0}
ight) 	imes 	ext{rate\_per\_hour}$$
  3. Return a formatted f-string where `hours`, `minutes`, and `seconds` are zero-padded to **2 digits** (`02d`) and `total_cost` is formatted to **2 decimal places** (`.2f`).
- **Worked Example**:
  - For `total_seconds = 8130` and `rate_per_hour = 50.0`:
    - `hours = 8130 // 3600 = 2`, remaining seconds `= 8130 % 3600 = 930`
    - `minutes = 930 // 60 = 15`, `seconds = 930 % 60 = 30`
    - `total_cost = (8130 / 3600.0) * 50.0 = 112.9166...`
    - **Expected Return String**: `"Duration: 02h:15m:30s | Billable: Rs. 112.92"`


````python
# =============================================================================
# PROBLEM 1 — INSTRUCTOR REFERENCE SOLUTION (5 Marks)
# =============================================================================

# --- PART A: Corrected Function (2 Marks) ---
def debug_power_calc(voltage_str, current_str, hours):
    # Fix 1: Explicitly cast numeric strings to float
    v = float(voltage_str)
    i = float(current_str)
    # Fix 2: Group denominator with parentheses so (2**0.5 * 2**0.5) = 2.0
    avg_power = (v * i) / ((2 ** 0.5) * (2 ** 0.5))
    energy_kwh = (avg_power * hours) / 1000.0
    # Fix 3: Use equality operator '==' instead of assignment '='
    is_zero = (hours == 0)
    status = f"Energy: {energy_kwh:.2f} kWh | Zero-Hours: {is_zero}"
    return round(avg_power, 2), round(energy_kwh, 4), status


# --- PART B: Completed Skeleton (3 Marks) ---
def decompose_duration_and_cost(total_seconds, rate_per_hour):
    '''
    Decomposes total_seconds into HH, MM, SS and computes billing summary string.
    '''
    hours = total_seconds // 3600
    rem = total_seconds % 3600
    minutes = rem // 60
    seconds = rem % 60

    total_cost = (total_seconds / 3600.0) * rate_per_hour
    return f"Duration: {hours:02d}h:{minutes:02d}m:{seconds:02d}s | Billable: Rs. {total_cost:.2f}"
````

````python
# PROBLEM 1 — SELF-CHECK VERIFICATION CELL
p_watts, e_kwh, msg = debug_power_calc("230.0", "4.0", 5)
assert p_watts == 460.0, f"Expected 460.0 W, got {p_watts}"
assert e_kwh == 2.3, f"Expected 2.3 kWh, got {e_kwh}"
assert msg == "Energy: 2.30 kWh | Zero-Hours: False", f"Unexpected status: {msg}"

summary = decompose_duration_and_cost(8130, 50.0)
assert summary == "Duration: 02h:15m:30s | Billable: Rs. 112.92", f"Mismatch: {summary}"
print("Problem 1 Verification Passed!")
````

## Problem 2: Lists, Slicing, Mutability & Tuple Unpacking (5 Marks)

### Problem Description & Scenario
A data acquisition unit collects sequential sensor readings in a Python `list`. Before computing statistics, the system must filter out extreme outliers (the single minimum and single maximum value) without altering the original raw list in memory, and extract specific sub-sequences and squared anomaly values for downstream diagnostics.

### Part A: Logical Flow Correction — Aliasing vs. Cloning & In-Place Sort Trap (2 Marks)
The function `debug_trim_outliers(raw_readings)` is required to:
1. Keep the caller's input list `raw_readings` **completely unchanged** in memory.
2. Create a sorted ascending copy of the readings.
3. Remove the smallest element (at index `0`) and the largest element (at index `-1`), and return the trimmed list.

**Defects to Fix in Part A**:
- **Bug 1 (List Aliasing)**: Writing `work_list = raw_readings` does not create a copy; both variables point to the same list in memory, so modifying `work_list` corrupts `raw_readings`. Use `.copy()` or `[:]`.
- **Bug 2 (`None` Return Trap)**: Writing `work_list = work_list.sort()` overwrites `work_list` with `None` because `list.sort()` sorts the list in-place and returns `None`.

### Part B: Skeleton Implementation (3 Marks)
Complete the function `analyze_sensor_batches(readings)`:
- **Input Parameter**:
  - `readings` (`list` of `int`/`float`): A sequence of at least 4 sensor readings (e.g., `[10, 55, 20, 60, 30, 40, 70, 80]`).
- **Required Operations**:
  1. `even_index_samples`: Use **step slicing** (`[::2]`) to extract elements located at even indices (`0, 2, 4, ...`).
  2. `reversed_last_four`: Use **slicing** to extract the last 4 elements of `readings` in **reverse order** (from the last element backward).
  3. `squared_anomalies`: Use a **list comprehension** to compute the square (`x ** 2`) of every number `x` in `readings` that is strictly greater than `50`.
  4. Return all three lists packed as a 3-element tuple: `(even_index_samples, reversed_last_four, squared_anomalies)`.
- **Worked Example**:
  - Input: `readings = [10, 55, 20, 60, 30, 40, 70, 80]`
  - `even_index_samples` $
ightarrow$ `[10, 20, 30, 70]`
  - `reversed_last_four` $
ightarrow$ `[80, 70, 40, 30]`
  - `squared_anomalies` $
ightarrow$ `[3025, 3600, 4900, 6400]` (squares of `55, 60, 70, 80`)


````python
# =============================================================================
# PROBLEM 2 — INSTRUCTOR REFERENCE SOLUTION (5 Marks)
# =============================================================================

# --- PART A: Corrected Function (2 Marks) ---
def debug_trim_outliers(raw_readings):
    # Fix 1: Clone the list using .copy() or slicing [:] so raw_readings is not mutated
    work_list = raw_readings.copy()
    # Fix 2: Call .sort() in-place without reassigning its None return value
    work_list.sort()
    work_list.pop(0)
    work_list.pop(-1)
    return work_list


# --- PART B: Completed Skeleton (3 Marks) ---
def analyze_sensor_batches(readings):
    '''
    Extracts sliced views and list comprehension results from readings.
    '''
    even_index_samples = readings[::2]
    reversed_last_four = readings[:-5:-1]  # or readings[-4:][::-1]
    squared_anomalies = [x ** 2 for x in readings if x > 50]

    return even_index_samples, reversed_last_four, squared_anomalies
````

````python
# PROBLEM 2 — SELF-CHECK VERIFICATION CELL
orig_data = [40, 10, 90, 30, 60, 20]
trimmed = debug_trim_outliers(orig_data)
assert orig_data == [40, 10, 90, 30, 60, 20], "Bug: original list was mutated!"
assert trimmed == [20, 30, 40, 60], f"Expected [20, 30, 40, 60], got {trimmed}"

evens, rev4, sq_anom = analyze_sensor_batches([10, 55, 20, 60, 30, 40, 70, 80])
assert evens == [10, 20, 30, 70], f"Even index slice error: {evens}"
assert rev4 == [80, 70, 40, 30], f"Reversed last 4 error: {rev4}"
assert sq_anom == [3025, 3600, 4900, 6400], f"List comprehension error: {sq_anom}"
print("Problem 2 Verification Passed!")
````

## Problem 3: Sets & Dictionary Data Structures (5 Marks)

### Problem Description & Scenario
An academic coordinator is analyzing student participation across two technical workshops (**NLP Track** and **Computer Vision Track**) where raw attendance logs contain duplicate entries. Simultaneously, the lab manager is consolidating hardware component inventories from two different storage rooms into a unified dictionary.

### Part A: Syntax & Logical Bug Correction (2 Marks)
The function `debug_workshop_audit(nlp_list, cv_list)` takes two lists of student roll numbers (`nlp_list` and `cv_list`) and is supposed to return a dictionary containing:
- `"both_tracks"`: The `set` of students who attended **both** workshops (Set Intersection).
- `"only_one_track"`: The `set` of students who attended **exactly one** of the two workshops, but not both (Set Symmetric Difference).

**Defects to Fix in Part A**:
- **Bug 1 (Empty Set Initialization)**: Writing `nlp_set = {}` creates an empty **dictionary** (`dict`), not a `set`, causing `AttributeError: 'dict' object has no attribute 'add'`.
- **Bug 2 (Invalid Set Operators)**: The code uses `nlp_set + cv_set` (which raises a `TypeError`) and `nlp_set - cv_set` (which computes set difference instead of intersection `&` and symmetric difference `^`).

### Part B: Skeleton Implementation (3 Marks)
Complete the function `merge_and_filter_inventory(warehouse_a, warehouse_b, min_stock)`:
- **Input Parameters**:
  - `warehouse_a` (`dict`): Maps component names (`str`) to stock counts (`int`), e.g., `{"ESP32": 10, "RPi4": 3, "Sensor": 15}`.
  - `warehouse_b` (`dict`): Maps component names (`str`) to stock counts (`int`), e.g., `{"ESP32": 5, "RPi4": 4, "Display": 12}`.
  - `min_stock` (`int`): Minimum combined stock threshold required to retain an item.
- **Required Operations**:
  1. Merge the counts from `warehouse_a` and `warehouse_b` into a new dictionary `combined`. If a component appears in both dictionaries, its combined count must be the **sum** of both counts. Use `.get(key, 0)` to avoid `KeyError`.
  2. Construct and return a dictionary `filtered` containing only those components from `combined` whose total quantity is $\ge 	ext{min\_stock}$.
- **Worked Example**:
  - With `warehouse_a = {"ESP32": 10, "RPi4": 3, "Sensor": 15}`, `warehouse_b = {"ESP32": 5, "RPi4": 4, "Display": 12}`, and `min_stock = 10`:
  - Combined counts: `{"ESP32": 15, "RPi4": 7, "Sensor": 15, "Display": 12}`
  - Filtered output ($\ge 10$): `{"ESP32": 15, "Sensor": 15, "Display": 12}` (`"RPi4": 7` is excluded).


````python
# =============================================================================
# PROBLEM 3 — INSTRUCTOR REFERENCE SOLUTION (5 Marks)
# =============================================================================

# --- PART A: Corrected Function (2 Marks) ---
def debug_workshop_audit(nlp_list, cv_list):
    # Fix 1: Use set() to initialize an empty set (not {})
    nlp_set = set()
    for s in nlp_list:
        nlp_set.add(s)
    cv_set = set(cv_list)
    # Fix 2: Use '&' for intersection and '^' for symmetric difference
    both = nlp_set & cv_set
    symmetric = nlp_set ^ cv_set
    return {"both_tracks": both, "only_one_track": symmetric}


# --- PART B: Completed Skeleton (3 Marks) ---
def merge_and_filter_inventory(warehouse_a, warehouse_b, min_stock):
    '''
    Combines stock counts from two warehouse dicts and filters by min_stock.
    '''
    combined = warehouse_a.copy()
    for item, qty in warehouse_b.items():
        combined[item] = combined.get(item, 0) + qty

    filtered = {item: qty for item, qty in combined.items() if qty >= min_stock}
    return filtered
````

````python
# PROBLEM 3 — SELF-CHECK VERIFICATION CELL
audit = debug_workshop_audit(["S1", "S2", "S3", "S2"], ["S2", "S3", "S4"])
assert audit["both_tracks"] == {"S2", "S3"}, f"Intersection error: {audit['both_tracks']}"
assert audit["only_one_track"] == {"S1", "S4"}, f"Symmetric diff error: {audit['only_one_track']}"

w_a = {"ESP32": 10, "RPi4": 3, "Sensor": 15}
w_b = {"ESP32": 5, "RPi4": 4, "Display": 12}
merged = merge_and_filter_inventory(w_a, w_b, min_stock=10)
assert merged == {"ESP32": 15, "Sensor": 15, "Display": 12}, f"Inventory merge error: {merged}"
print("Problem 3 Verification Passed!")
````

## Problem 4: Control Flow, Loops & Numerical Algorithms (5 Marks)

### Problem Description & Scenario
This problem tests your mastery of conditional branching (`if-elif-else`), loop control statements (`break` and `continue`), mathematical digit extraction, and dictionary traversal.

### Part A: Logical Flow Bug Correction — Premature Return in Prime Check (2 Marks)
A prime number is an integer $n > 1$ that has no positive divisors other than $1$ and itself.
In `debug_is_prime(n)` below, the programmer placed `else: return True` **inside** the `for` loop body! Because of this indentation error, when testing a composite odd number like `n = 9`, the first iteration checks `9 % 2 == 0` (which is `False`) and immediately jumps to `else: return True` without ever testing `d = 3`! Furthermore, it fails to handle boundary inputs where `n <= 1`.

**Defects to Fix in Part A**:
- Return `False` immediately for any `n <= 1`.
- Ensure the loop checks all potential divisors $d$ from $2$ up to $\lfloor\sqrt{n}
floor$ (or $n - 1$) before concluding that `n` is prime (`return True` must be outside the `for` loop).

### Part B: Skeleton Implementation (3 Marks)
Complete the function `filter_armstrong_and_grade_summary(numbers, student_marks_dict)`:
- **Input Parameters**:
  - `numbers` (`list` of `int`): A stream of integers (e.g., `[-15, 0, 153, 120, 370, 9474, 9999, 407]`).
  - `student_marks_dict` (`dict`): Maps student names (`str`) to numeric scores (`int`/`float`).
- **Required Operations**:
  1. **Stream Filtering & Armstrong Number Detection**:
     - Iterate through `numbers` in order:
       - If a number is $\le 0$, skip it immediately using `continue`.
       - If the sentinel number `9999` is encountered, stop scanning the list immediately using `break` (do not check `9999` or any numbers after it).
       - For each valid positive number, determine whether it is an **Armstrong number** (a $k$-digit integer equal to the sum of its digits each raised to the $k$-th power, such as $153 = 1^3 + 5^3 + 3^3 = 153$ or $9474 = 9^4 + 4^4 + 7^4 + 4^4 = 9474$). Append all detected Armstrong numbers to `armstrong_found`.
  2. **Student Grade Classification**:
     - Iterate over `student_marks_dict.items()` and populate `grade_dict` mapping each student's name to:
       - `"Distinction"` if `score >= 75`
       - `"Pass"` if `50 <= score < 75`
       - `"Fail"` if `score < 50`
  3. Return the tuple `(armstrong_found, grade_dict)`.
- **Worked Example**:
  - `numbers = [-15, 0, 153, 120, 370, 9474, 9999, 407]` $
ightarrow$ skips `-15` and `0`, keeps `153`, `370`, `9474`, breaks at `9999` (ignoring `407`) $
ightarrow$ `armstrong_found = [153, 370, 9474]`.


````python
# =============================================================================
# PROBLEM 4 — INSTRUCTOR REFERENCE SOLUTION (5 Marks)
# =============================================================================

# --- PART A: Corrected Function (2 Marks) ---
def debug_is_prime(n):
    if n <= 1:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            return False
    # Return True ONLY after the entire loop completes without finding a factor!
    return True


# --- PART B: Completed Skeleton (3 Marks) ---
def filter_armstrong_and_grade_summary(numbers, student_marks_dict):
    '''
    1. Finds Armstrong numbers in 'numbers' (ignoring <=0, stopping at 9999).
    2. Classifies student_marks_dict into Distinction (>=75), Pass (50-74), Fail (<50).
    '''
    armstrong_found = []
    for num in numbers:
        if num <= 0:
            continue
        if num == 9999:
            break
        # Check Armstrong condition:
        k = len(str(num))
        temp = num
        power_sum = 0
        while temp > 0:
            power_sum += (temp % 10) ** k
            temp //= 10
        if power_sum == num:
            armstrong_found.append(num)

    grade_dict = {}
    for name, score in student_marks_dict.items():
        if score >= 75:
            grade_dict[name] = "Distinction"
        elif score >= 50:
            grade_dict[name] = "Pass"
        else:
            grade_dict[name] = "Fail"

    return armstrong_found, grade_dict
````

````python
# PROBLEM 4 — SELF-CHECK VERIFICATION CELL
assert debug_is_prime(1) == False
assert debug_is_prime(2) == True
assert debug_is_prime(9) == False, "Bug: 9 is composite (3x3), not prime!"
assert debug_is_prime(29) == True

nums_input = [-15, 0, 153, 120, 370, 9474, 9999, 407]
marks_input = {"Anu": 82, "Binu": 64, "Chitra": 45}
arms, grades = filter_armstrong_and_grade_summary(nums_input, marks_input)
assert arms == [153, 370, 9474], f"Expected [153, 370, 9474] (stopping before 407), got {arms}"
assert grades == {"Anu": "Distinction", "Binu": "Pass", "Chitra": "Fail"}, f"Grade error: {grades}"
print("Problem 4 Verification Passed!")
````

## Problem 5: User-Defined Functions, Arguments & Variable Scope (5 Marks)

### Problem Description & Scenario
Modular Python programs rely on default arguments, variable-length argument packing (`*args`), and nested closures with enclosing scope variables (`nonlocal`) to maintain state cleanly without polluting the global namespace.

### Part A: Logical Bug Correction — The Mutable Default Argument Trap (2 Marks)
The function `debug_append_log(event, tag="INFO", log_history=[])` is intended to append a formatted log string `f"[{tag}] {event}"` to an existing `log_history` list if one is passed, or create and return a brand-new 1-element list if `log_history` is omitted.

**Defect to Fix in Part A**:
- In Python, default argument values are evaluated **only once** when `def` is executed, not each time the function is called! Using a mutable list `log_history=[]` as a default argument causes every call that omits `log_history` to share and mutate the **same persistent list** across calls!
- Fix the function using the standard Python idiom (`log_history=None`, initializing `log_history = []` inside the function when `log_history is None`).

### Part B: Skeleton Implementation (3 Marks)
Complete the two functions below:
1. `weighted_sum_of_squares(scale=1.0, *numbers)`:
   - **Input Parameters**:
     - `scale` (`float`, default `1.0`): Multiplier applied to the sum of squares.
     - `*numbers`: An arbitrary number of numeric positional arguments (could be 0, 1, or many numbers).
   - **Behavior**: Compute the sum of squares of all values in `numbers` ($\sum x_i^2$), multiply by `scale`, and return the result rounded to 4 decimal places. If no numbers are passed after `scale`, return `0.0`.
   - **Worked Example**: `weighted_sum_of_squares(2.0, 3, 4)` computes $2.0 	imes (3^2 + 4^2) = 2.0 	imes 25 = 50.0$.
2. `make_step_accumulator(initial_value=0, step=1)`:
   - **Input Parameters**: `initial_value` (`int`/`float`), `step` (`int`/`float`).
   - **Behavior**: Stores `current = initial_value` in the enclosing scope and defines a nested inner function `step_up()`. Each time `step_up()` is called, it must use the `nonlocal` keyword to add `step` to `current` and return the new value of `current`.
   - **Worked Example**: If `acc = make_step_accumulator(10, 5)`, then calling `acc()` three times returns `15`, then `20`, then `25`.


````python
# =============================================================================
# PROBLEM 5 — INSTRUCTOR REFERENCE SOLUTION (5 Marks)
# =============================================================================

# --- PART A: Corrected Function (2 Marks) ---
def debug_append_log(event, tag="INFO", log_history=None):
    if log_history is None:
        log_history = []
    log_history.append(f"[{tag}] {event}")
    return log_history


# --- PART B: Completed Skeleton (3 Marks) ---
def weighted_sum_of_squares(scale=1.0, *numbers):
    '''
    Computes scale * sum(x^2 for x in numbers).
    '''
    return round(scale * sum(x ** 2 for x in numbers), 4)


def make_step_accumulator(initial_value=0, step=1):
    '''
    Returns a closure function that increments initial_value by step on each call.
    '''
    current = initial_value

    def step_up():
        nonlocal current
        current += step
        return current

    return step_up
````

````python
# PROBLEM 5 — SELF-CHECK VERIFICATION CELL
log1 = debug_append_log("System Boot")
log2 = debug_append_log("Sensor Online", tag="WARN")
assert log1 == ["[INFO] System Boot"], f"Unexpected log1: {log1}"
assert log2 == ["[WARN] Sensor Online"], f"Mutable default bug persisted: {log2}"

assert weighted_sum_of_squares(2.0, 3, 4) == 50.0, "Expected 2.0 * (9 + 16) = 50.0"
assert weighted_sum_of_squares(1.0) == 0.0, "Empty *numbers should return 0.0"

acc = make_step_accumulator(10, 5)
assert acc() == 15
assert acc() == 20
assert acc() == 25
print("Problem 5 Verification Passed!")
````

## Problem 6: String Processing, File I/O & Exception Handling (5 Marks)

### Problem Description & Scenario
An environmental telemetry logger writes sensor records to text files and parses incoming CSV files uploaded from remote weather stations. Because remote transmissions are prone to noise and missing files, your file ingestion logic must use context managers (`with open(...)`), string methods (`.strip()`, `.split()`), and targeted exception handlers (`FileNotFoundError`, `ValueError`).

### Part A: File Mode & Exception Handling Bug Correction (2 Marks)
Inspect the two utility functions `debug_append_record(filepath, record_line)` and `debug_read_safe(filepath)`:

Fix both functions so they behave as specified.

### Part B: Skeleton Implementation (3 Marks)
Complete the function `parse_dirty_csv_file(filepath)`:

  2. Skip the first header line (`f.readline()`).
  3. Iterate through the remaining lines:

  5. Return the 3-element tuple `(valid_count, mean_val, corrupted_ids)`.
- **Worked Example**:
  - For the sample file above, `S1` (`10.5`), `S3` (`20.5`), and `S5` (`30.0`) are valid (`valid_count = 3`, `mean_val = 20.33`), while `S2` and `S4` fail float conversion (`corrupted_ids = ["S2", "S4"]`).
  - **Expected Return**: `(3, 20.33, ["S2", "S4"])`.


````python
# =============================================================================
# PROBLEM 6 — INSTRUCTOR REFERENCE SOLUTION (5 Marks)
# =============================================================================

# --- PART A: Corrected Functions (2 Marks) ---
def debug_append_record(filepath, record_line):
    # Fix 1: Open in 'a' (append) mode so existing lines are preserved
    with open(filepath, "a", encoding="utf-8") as f:
        f.write(record_line.strip() + "\n")

def debug_read_safe(filepath):
    # Fix 2: Catch FileNotFoundError specifically
    try:
        with open(filepath, "r", encoding="utf-8") as f:
            return f.read()
    except FileNotFoundError:
        return "FILE_MISSING"


# --- PART B: Completed Skeleton (3 Marks) ---
def parse_dirty_csv_file(filepath):
    '''
    Parses a CSV file with header 'ID,Reading', handling FileNotFoundError
    and per-row ValueError exceptions.
    '''
    valid_readings = []
    corrupted_ids = []

    try:
        with open(filepath, "r", encoding="utf-8") as f:
            header = f.readline()  # Skip header line
            for line in f:
                cleaned = line.strip()
                if not cleaned:
                    continue
                parts = cleaned.split(",")
                if len(parts) < 2:
                    continue
                sid, raw_val = parts[0].strip(), parts[1].strip()
                try:
                    val = float(raw_val)
                    valid_readings.append(val)
                except ValueError:
                    corrupted_ids.append(sid)
    except FileNotFoundError:
        return 0, 0.0, []

    count = len(valid_readings)
    mean_val = round(sum(valid_readings) / count, 2) if count > 0 else 0.0
    return count, mean_val, corrupted_ids
````

````python
# PROBLEM 6 — SELF-CHECK VERIFICATION CELL
test_log = "midsem_test_log.txt"
with open(test_log, "w", encoding="utf-8") as f:
    f.write("Line1\n")
debug_append_record(test_log, "Line2")
assert debug_read_safe(test_log) == "Line1\nLine2\n", "Append mode bug!"
assert debug_read_safe("non_existent_midsem_file.txt") == "FILE_MISSING", "Exception handling bug!"

csv_file = "midsem_sensors.csv"
with open(csv_file, "w", encoding="utf-8") as f:
    f.write("ID,Reading\nS1,10.5\nS2,ERR_TIMEOUT\nS3,20.5\nS4,NULL\nS5,30.0\n")

cnt, avg_v, bad_ids = parse_dirty_csv_file(csv_file)
assert cnt == 3, f"Expected 3 valid rows, got {cnt}"
assert avg_v == 20.33, f"Expected mean 20.33, got {avg_v}"
assert bad_ids == ["S2", "S4"], f"Expected ['S2', 'S4'], got {bad_ids}"
assert parse_dirty_csv_file("missing_csv.csv") == (0, 0.0, [])
print("Problem 6 Verification Passed!")
````

## Problem 7: Object-Oriented Programming — Classes, Encapsulation & Inheritance (5 Marks)

### Problem Description & Scenario
You are building an Object-Oriented resource management system for an AI research laboratory. The system tracks active GPU compute nodes using a shared class variable and models project funding accounts using inheritance, `@property` encapsulation, and method overriding.

### Part A: Syntax & Class Variable Shadowing Correction (2 Marks)
Inspect the class `ComputeNode` below:
1. It defines a shared class variable `active_nodes = 0` intended to count how many `ComputeNode` instances have been created across the entire program.
2. In `__init__(self, node_id, gpus)`, writing `self.active_nodes += 1` does **not** update `ComputeNode.active_nodes`; instead, it creates an instance variable `self.active_nodes = 1` that shadows the class variable, leaving `ComputeNode.active_nodes` stuck at `0`!
3. The instance method `get_node_info()` is missing the mandatory `self` parameter in its signature, causing a `TypeError` when called on an instance (`n1.get_node_info()`).

Fix both bugs in `ComputeNode`.

### Part B: Skeleton Implementation (3 Marks)
Complete the base class `ResearchAccount` and derived subclass `AIGrantAccount(ResearchAccount)`:
1. **Base Class `ResearchAccount`**:
   - `__init__(self, account_id, holder_name, balance)`: Initialize `self.account_id = account_id`, `self.holder_name = holder_name`, and the protected attribute `self._balance = float(balance)`.
   - `@property def balance(self)`: Getter method that returns `self._balance`.
   - `@balance.setter def balance(self, new_val)`: Setter method that validates `new_val`. If `new_val < 0`, raise a `ValueError("Balance cannot be negative")`; otherwise, update `self._balance = float(new_val)`.
   - `compute_annual_budget(self)`: Return `self._balance`.
2. **Derived Subclass `AIGrantAccount(ResearchAccount)`**:
   - `__init__(self, account_id, holder_name, balance, gpu_subsidy)`: Call the parent constructor `super().__init__(account_id, holder_name, balance)` and store `self.gpu_subsidy = float(gpu_subsidy)`.
   - `compute_annual_budget(self)`: **Override** the parent method to return the sum of the parent's `compute_annual_budget()` (via `super()`) plus `self.gpu_subsidy`.
- **Worked Example**:
  - `grant = AIGrantAccount("GR-101", "Dr. Sunil", 500000.0, 150000.0)`
  - `grant.compute_annual_budget()` $
ightarrow$ returns `650000.0` (`500000.0 + 150000.0`).
  - `grant.balance = -500.0` $
ightarrow$ raises `ValueError`.


````python
# =============================================================================
# PROBLEM 7 — INSTRUCTOR REFERENCE SOLUTION (5 Marks)
# =============================================================================

# --- PART A: Corrected Class (2 Marks) ---
class ComputeNode:
    active_nodes = 0  # Shared class variable

    def __init__(self, node_id, gpus):
        self.node_id = node_id
        self.gpus = gpus
        # Fix 1: Increment via Class Name so all instances share the counter
        ComputeNode.active_nodes += 1

    # Fix 2: Include 'self' as first parameter of instance method
    def get_node_info(self):
        return f"Node {self.node_id} ({self.gpus} GPUs)"


# --- PART B: Completed OOP Hierarchy (3 Marks) ---
class ResearchAccount:
    def __init__(self, account_id, holder_name, balance):
        self.account_id = account_id
        self.holder_name = holder_name
        self._balance = float(balance)

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, new_val):
        if new_val < 0:
            raise ValueError("Balance cannot be negative")
        self._balance = float(new_val)

    def compute_annual_budget(self):
        return self._balance


class AIGrantAccount(ResearchAccount):
    def __init__(self, account_id, holder_name, balance, gpu_subsidy):
        super().__init__(account_id, holder_name, balance)
        self.gpu_subsidy = float(gpu_subsidy)

    def compute_annual_budget(self):
        return super().compute_annual_budget() + self.gpu_subsidy
````

````python
# PROBLEM 7 — SELF-CHECK VERIFICATION CELL
ComputeNode.active_nodes = 0
n1 = ComputeNode("DGX-01", 8)
n2 = ComputeNode("DGX-02", 4)
assert ComputeNode.active_nodes == 2, f"Class variable bug: got {ComputeNode.active_nodes}"
assert n1.get_node_info() == "Node DGX-01 (8 GPUs)"

grant = AIGrantAccount("GR-101", "Dr. Sunil", 500000.0, 150000.0)
assert isinstance(grant, ResearchAccount), "AIGrantAccount must inherit from ResearchAccount"
assert grant.compute_annual_budget() == 650000.0, f"Overridden method error: {grant.compute_annual_budget()}"
grant.balance = 600000.0
assert grant.compute_annual_budget() == 750000.0

try:
    grant.balance = -500.0
    assert False, "Setter should have raised ValueError for negative balance!"
except ValueError:
    pass
print("Problem 7 Verification Passed!")
````
