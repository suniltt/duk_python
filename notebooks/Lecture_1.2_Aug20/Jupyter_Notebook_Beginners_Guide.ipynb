## 1. How Jupyter Works: The Client-Kernel Architecture

Unlike traditional text editors (where you write a script and run it from top to bottom all at once), Jupyter uses an **interactive client-server model**:

```text
  +-------------------------------------------------------------+
  |                    Web Browser (Frontend)                   |
  |   Displays Cells, Code, Markdown, Tables, Images, Charts    |
  +------------------------------+------------------------------+
                                 | (ZeroMQ / WebSockets)
                                 v
  +-------------------------------------------------------------+
  |                Jupyter Server (Background Daemon)           |
  |   Manages Files, Routes Communication, Spawns Terminals     |
  +------------------------------+------------------------------+
                                 |
                                 v
  +-------------------------------------------------------------+
  |                 IPython Kernel (The Brain)                  |
  |   Executes Python code, holds memory state and variables    |
  +-------------------------------------------------------------+
```

### 💡 Key Takeaways:
* When you type code in a cell and run it, the code is sent to the **Kernel**.
* The **Kernel retains all variables and functions in memory** until you restart it or close the session.
* If you close your browser tab, your Kernel is **still running** in the background on the server!

## 3. Modal Editing: Command Mode vs. Edit Mode

Jupyter has **two distinct modes**, inspired by the Vim editor. Mastering the difference between them is the secret to fast coding:

```text
                   Press [Enter] or Click Inside
      +--------------------------------------------------------+
      |                                                        |
      v                                                        |
  [ EDIT MODE ]                                          [ COMMAND MODE ]
  Border: Green or Blue line                             Border: Grey or Blue box
  Cursor: Blinking inside cell                           Cursor: None (Cell is selected)
  Action: Typing code or text                            Action: Managing cells
      |                                                        ^
      |              Press [Esc] or Click Outside              |
      +--------------------------------------------------------+
```

👉 **Action Check:** Look at the border of this cell. Press **`Enter`** to enter Edit mode, then press **`Esc`** to return to Command mode!

## 4. The Essential Keyboard Shortcuts Cheat Sheet

### ⚡ Running Cells (Works in Both Modes):
* **`Shift + Enter`**: Run the current cell and **move to the next cell** (most common).
* **`Ctrl + Enter`** (or `Cmd + Enter` on Mac): Run the current cell and **stay on it** (great for testing).
* **`Alt + Enter`** (or `Option + Enter`): Run the current cell and **insert a new blank cell immediately below**.

### 📦 Managing Cells (In Command Mode — Press `Esc` First):
* **`A`**: Insert a new cell **Above** the selected cell.
* **`B`**: Insert a new cell **Below** the selected cell.
* **`D, D`** (Press `D` twice quickly): **Delete** the selected cell.
* **`Z`**: **Undo** cell deletion.
* **`C`**: Copy selected cell.
* **`X`**: Cut selected cell.
* **`V`**: Paste cell below.
* **`M`**: Convert cell to **Markdown** (text/notes).
* **`Y`**: Convert cell to **Code** (Python).
* **`1` to `6`**: Turn cell into Markdown Heading 1 (`#`) through Heading 6 (`######`).
* **`L`**: Toggle **Line Numbers** in the code cell.

## 6. Comprehensive Markdown Tutorial for Beginners

To create or edit a Markdown cell in Jupyter:
1. Select a cell and press **`Esc`** then **`M`** (or choose *Markdown* from the toolbar dropdown).
2. Type your formatted text.
3. Press **`Shift + Enter`** to render it into beautiful styled text.
4. **Double-click** on any rendered Markdown cell to reopen and edit its raw source code!

## 7. Understanding the Execution State & Execution Numbers

Notice the brackets `In [ ]:` to the left of every code cell. These indicators tell you the exact state of the Kernel:

| Indicator | Meaning | Action / State |
| :--- | :--- | :--- |
| `In [ ]:` | **Not Yet Executed** | The cell has not been run since the kernel started. |
| `In [*]:` | **Currently Running (Busy)** | The kernel is actively computing this cell. Wait for it to finish! |
| `In [4]:` | **Executed (Order = 4)** | This was the 4th cell executed in the current session. |

### ⚠️ The #1 Jupyter Pitfall: Out-of-Order Execution
In a script, code runs top-to-bottom. In Jupyter, **you can run cells in any order** (e.g., run cell 5, then cell 2, then cell 5 again).

```text
  Cell A:  x = 10         [Executed 1st  -> In [1]]
  Cell B:  x = 50         [Executed 3rd  -> In [3]]
  Cell C:  print(x)       [Executed 2nd  -> Printed 10, but now in memory x is 50!]
```

> **🛡️ Best Practice Rule:** Before submitting an assignment or sharing a research notebook, always click **`Kernel -> Restart Kernel and Run All Cells`** to verify that your notebook executes cleanly from top to bottom.

## 8. Managing the Kernel: Interrupting & Selecting Kernels

### 🛑 What to do when code is stuck in an infinite loop?
If you accidentally write an infinite loop, your cell will show `In [*]:` and refuse to stop.
* **Solution:** Click the **Square Stop Button (■)** on the toolbar or press **`Kernel -> Interrupt Kernel`** (or press `I, I` in Command Mode).

### 🔄 What to do when memory is corrupted or cluttered?
* Click the **Circular Arrow Button (↻)** or select **`Kernel -> Restart Kernel`**. This wipes the RAM clean so you can start fresh.

### 🎯 How to Choose / Switch Python Kernels (Virtual Environments)?
Look at the **top-right corner** of your Jupyter window:
1. Click on the kernel name (e.g., `Python 3 (ipykernel)` or `Python [conda env:base]`).
2. A dialog opens listing all available Python virtual environments on your machine.
3. Select the environment tailored for your course (e.g., `Python (ai_env)`).

## 9. IPython Magic Commands: Supercharging Productivity

Magic commands are special prefixes provided by IPython that give you superpowers inside notebook cells:

### 1. Line Magics (Prefixed with a single `%` — applies to one line):
* `%timeit <expr>`: Benchmark code execution speed.
* `%whos`: List all variables in memory with their types and sizes.
* `%pwd`: Print current working directory.
* `%ls`: List files in the current folder.
* `%load <filename.py>`: Load code from a file directly into a cell.

### 2. Cell Magics (Prefixed with `%%` — applies to the entire cell):
* `%%timeit`: Benchmark the whole cell.
* `%%writefile script.py`: Save the contents of the cell directly to a file on disk.
* `%%bash`: Run the cell as a bash shell script.

Let's test these magics below:

`````python
# Example 1: Timing code with %timeit
%timeit sum(x * x for x in range(10_000))

# Example 2: Inspecting memory with %whos
%whos
`````

### 3. Shell Commands with `!`:
Prefix any terminal command with an exclamation mark `!` to run it directly from a code cell!

`````python
# Checking installed python version and listing directory contents via shell
!python --version
!pwd
`````
