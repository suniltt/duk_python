# Part 1: Python Core Data Types & Memory Object Model


In Python, **variables are not typed memory boxes** (as in C/C++ or Java). Instead, **variables are reference tags (labels)** bound to objects located in memory.

```
       Variable Name (Label)                  Object in Memory
          +-------------+                     +-----------------------+
          |      x      | ----------------->  | Type  : int           |
          +-------------+                     | Value : 42            |
                                              | Id    : 0x7fff489b    |
                                              +-----------------------+
```

### Python's Five Fundamental Atomic Types:
1. **`int` (Integer)**: Whole numbers. In Python 3, integers have **arbitrary precision** (limited only by available system RAM).
2. **`float` (Floating-Point)**: Real numbers implemented as IEEE 754 double-precision floats (64-bit).  ( Question .What is EEE 754 double-precision floats)
3. **`complex` (Complex Number)**: Numbers with real and imaginary parts ($a + bj$, where $j = \sqrt{-1}$).  
4. **`bool` (Boolean)**: Truth values (`True` or `False`). A direct subclass of `int`.
5. **`str` (String)**: Immutable sequences of Unicode characters.  ( Q: What is immutable)


### 📋 Rules for Legal Variable Names:
1. Must begin with a letter (`a-z`, `A-Z`) or an underscore (`_`).
2. Cannot begin with a number (e.g., `2nd_sensor` is **illegal**; `sensor_2` is **legal**).
3. Can only contain alphanumeric characters and underscores (`A-z`, `0-9`, and `_`).
4. Case-sensitive: `learning_rate`, `Learning_Rate`, and `LEARNING_RATE` are three distinct entities.
5. Cannot use Python reserved keywords (e.g., `for`, `class`, `import`, `def`, `lambda`, `pass`).

````python
# 1.1 Integer: Arbitrary Precision (No 32-bit or 64-bit integer overflow!)
normal_int = 100
large_int = 2 ** 100  # A 31-digit integer!
astronomical_int = 10 ** 50

print("Normal int:", normal_int)
print("2^100      :", large_int)
print("10^50      :", astronomical_int)
print("Type of astronomical_int:", type(astronomical_int))

# Different Integer Number Bases:
bin_val = 0b101101   # Binary prefix: 0b
oct_val = 0o755      # Octal prefix: 0o
hex_val = 0xDEADBEEF # Hexadecimal prefix: 0x

print(f"Bases in Decimal: binary={bin_val}, octal={oct_val}, hex={hex_val}")
````

````python
# 1.2 Floating-Point: Representation and Precision Limits
pi_val = 3.141592653589793
avogadro = 6.02214076e23  # Scientific notation: 6.02214076 * 10^23
planck = 6.62607015e-34    # Scientific notation: 6.62607015 * 10^-34

print("Pi approximation :", pi_val)
print("Avogadro constant:", avogadro)
print("Planck constant  :", planck)

````

````python
# 1.3 Complex Numbers: Native support for Science and Signal Processing
z1 = 4 + 3j
z2 = 2 - 5j

print("z1 =", z1)
print("z1 Real part      :", z1.real)
print("z1 Imaginary part :", z1.imag)
print("Conjugate of z1   :", z1.conjugate())
print("z1 + z2           :", z1 + z2)
print("z1 * z2           :", z1 * z2)
print("Magnitude |z1|    :", abs(z1))  # sqrt(4^2 + 3^2) = 5.0
````

# Part 2: Type Inspection, Type Conversion & Encoding


Python is **strongly typed** (it enforces type compatibility and will not automatically turn `"5"` into `5` when adding) and **dynamically typed** (variable labels can be rebound to different types).

### 1. Type Conversion Taxonomy:
- **Implicit Conversion (Type Coercion)**: Python automatically promotes smaller types to larger types in mixed expressions (e.g. `int + float` $\rightarrow$ `float`).
- **Explicit Conversion (Type Casting)**: The programmer manually invokes conversion functions:
  - `int(x, [base])`: Converts number or string to integer. Truncates floats towards zero.
  - `float(x)`: Converts integer or numeric string to float.
  - `str(x)`: Converts any object into its string representation.
  - `bool(x)`: Evaluates the **truthiness** of an object.
  - `complex(real, [imag])`: Constructs a complex number.

### 2. Character Encodings: `ord()` and `chr()`  
( skip this if you dont understand)
- `ord(char)`: Returns the integer ASCII / Unicode code point of a single character. (q: What is ASCII /Unicode)
- `chr(code)`: Returns the string character corresponding to the integer code point.


````python
# 2.1 Implicit vs. Explicit Type Conversion
num_int = 25
num_float = 4.75

# Implicit widening (Coercion):
mixed_sum = num_int + num_float
print("num_int + num_float =", mixed_sum, "| Type:", type(mixed_sum))

# Explicit Type Casting:
# Truncation: int() DOES NOT round; it discards the fractional portion
print("int(4.75)   =", int(4.75))
print("int(-4.75)  =", int(-4.75))  # Truncates towards zero -> -4

# String to number conversion:
str_val = "1024"
parsed_int = int(str_val)
print(f"String '{str_val}' parsed to int: {parsed_int}, multiplied: {parsed_int * 2}")

# Parsing different number bases using int(string, base):
print("Hex '1A' to int     :", int("1A", 16))
print("Binary '1101' to int:", int("1101", 2))
````

````python
# 2.2 Character Encoding: ord() and chr()
char1 = 'A'
char2 = 'a'

code1 = ord(char1)
code2 = ord(char2)

print(f"Character: '{char1}' -> ASCII code: {code1}")
print(f"Character: '{char2}' -> ASCII code: {code2}")

````

Formatted String Literals (f-strings)


An **f-string** (Formatted String Literal) is the standard, most readable, and fastest way to format strings in modern Python.


Prefix the string literal with an `f` or `F`, and place variables or expressions inside curly braces `{}`:
```python
name = "Sunil"
print(f"Hello, {name}!")

````python
# Example 1: Basic variable substitution
# f-strings allow embedding variable values directly into a string.
name = "Alice"
age = 30
print(f"My name is {name} and I am {age} years old.")

# Example 2: Expression evaluation inside f-string
# You can include Python expressions directly within the curly braces {}.
x = 10
y = 5
print(f"The sum of {x} and {y} is {x + y}.")
print(f"The product of {x} and {y} is {x * y}.")

# Example 3: Formatting numbers (precision, alignment)
# f-strings support powerful formatting options using a colon ':' followed by format specifiers.
pi = 3.14159265
data_size = 123456789
percentage = 0.85

# :.2f formats a float to two decimal places.
print(f"Pi with 2 decimal places: {pi:.2f}")
# :.4f formats a float to four decimal places.
print(f"Pi with 4 decimal places: {pi:.4f}")
# :, adds a comma as a thousands separator for numbers.
print(f"Data size formatted: {data_size:,}")
# :.1% formats a float as a percentage with one decimal place (multiplies by 100).
print(f"Percentage: {percentage:.1%}")

# Example 4: Alignment and padding
# You can specify field width and alignment within f-strings.
item = "Laptop"
price = 1200.50
quantity = 2

# <10: left-aligns string within a field of 10 characters.
# >8: right-aligns string within a field of 8 characters.
print(f"{'Item':<10} | {'Price':>8} | {'Qty':>4}")
# :-^26: center-aligns '-' within a field of 26 characters.
print(f"{'':-^26}") # Separator line
# :<10: left-aligns 'item'. :>8.2f: right-aligns 'price' with 2 decimal places.
# :>4d: right-aligns 'quantity' as an integer in a field of 4 characters.
print(f"{item:<10} | {price:>8.2f} | {quantity:>4d}")

# Example 5: Calling functions inside f-strings
# You can call functions or methods directly within the curly braces.
def greet(person):
    return f"Hello, {person}"

print(f"Calling a function: {greet('Bob')}")
````

# Part 4: Arithmetic Operators & Computational Mechanics


Python features 7 fundamental arithmetic operators.

### Detailed Operator Breakdown:
| Operator | Symbol | Example ($a=19, b=4$) | Evaluation | Important Behavioural Note |
| :--- | :---: | :---: | :---: | :--- |
| **Addition** | `+` | `19 + 4` | `23` | Commutative; also concatenates sequences |
| **Subtraction** | `-` | `19 - 4` | `15` | Also unary negation `-x` |
| **Multiplication** | `*` | `19 * 4` | `76` | Commutative; also repeats sequences |
| **True Division** | `/` | `19 / 4` | `4.75` | **Always returns a `float`** in Python 3 |
| **Floor Division** | `//` | `19 // 4` | `4` | Rounds down to the nearest floor integer towards $-\infty$ |
| **Modulus (Remainder)** | `%` | `19 % 4` | `3` | Remainder: $a - (b \times (a // b))$ |
| **Exponentiation** | `**` | `2 ** 4` | `16` | Power operator; **right-to-left associativity** |

### Critical Difference: True Division (`/`) vs Floor Division (`//`)
- `/` always yields a `float` (even for `4 / 2` $\rightarrow$ `2.0`).
- `//` discards the fractional part by rounding towards negative infinity ($-\infty$).
  - `7 // 2` evaluates to `3`
  - `-7 // 2` evaluates to `-4` (NOT `-3`, because $-3.5$ rounded down toward $-\infty$ is $-4$!).


````python
# 4.1 True Division vs Floor Division vs Modulus
x, y = 19, 4

print(f"x = {x}, y = {y}")
print("x / y  =", x / y, " -> True division always returns float")
print("x // y =", x // y, " -> Floor division (integer quotient)")
print("x % y  =", x % y, " -> Modulus (remainder)")

# Mathematical Invariant in Python: x == (y * (x // y)) + (x % y)
invariant_holds = (x == (y * (x // y)) + (x % y))
print("Division Invariant verified:", invariant_holds)

# divmod() built-in function returns (quotient, remainder) simultaneously:
q, r = divmod(x, y)
print(f"divmod({x}, {y}) = (Quotient={q}, Remainder={r})")
````

````python
# 4.2 Negative Number Floor Division and Modulo
# This section demonstrates how floor division (//) and the modulo operator (%) behave
# with negative numbers in Python, which can be a common source of confusion for
# programmers familiar with other languages like C or Java.

# Perform floor division for a positive dividend.
# 17 // 5 = 3 (since 5 * 3 = 15, and 5 * 4 = 20, 15 is the largest multiple of 5 <= 17)
pos_div = 17 // 5

# Perform floor division for a negative dividend.
# -17 // 5 = -4. Python's floor division rounds towards negative infinity.
# The closest multiple of 5 less than or equal to -17 is -20 (5 * -4 = -20).
neg_div = -17 // 5

# Calculate the modulo (remainder) for a positive dividend.
# 17 % 5 = 2 (since 17 = 5 * 3 + 2)
pos_mod = 17 % 5

# Calculate the modulo (remainder) for a negative dividend.
# -17 % 5 = 3. The sign of the remainder in Python matches the sign of the divisor.
# This can be verified by the invariant: x == (y * (x // y)) + (x % y).
# So, -17 == (5 * (-4)) + 3 => -17 == -20 + 3 => -17 == -17.
neg_mod = -17 % 5

print("17 // 5  =", pos_div)
print("-17 // 5 =", neg_div, " (Floored towards -infinity, NOT zero!)")
print("17 % 5   =", pos_mod)
print("-17 % 5  =", neg_mod, " (Notice: 5 * (-4) + 3 = -17!)")
````

````python
# 4.3 Exponentiation (**) and Right-to-Left Associativity
# When multiple ** operators appear, evaluation proceeds from RIGHT to LEFT!
val_eval = 2 ** 3 ** 2
print("2 ** 3 ** 2 =", val_eval)
print("Equivalent to 2 ** (3 ** 2) = 2 ** 9 =", 2 ** (3 ** 2))
print("NOT equivalent to (2 ** 3) ** 2 =", (2 ** 3) ** 2)

# Fractional exponents compute roots:
square_root_64 = 64 ** 0.5
cube_root_27 = 27 ** (1/3)
print("64 ** 0.5     =", square_root_64)
print("27 ** (1/3)   =", cube_root_27)
````

````python
# 4.4 Compound Assignment Operators
acc = 100
print("Initial Accumulator:", acc)

acc += 25    # acc = acc + 25
print("After += 25 :", acc)

acc -= 15    # acc = acc - 15
print("After -= 15 :", acc)

acc *= 2     # acc = acc * 2
print("After *= 2  :", acc)

acc //= 7    # acc = acc // 7
print("After //= 7 :", acc)

acc %= 5     # acc = acc % 5
print("After %= 5  :", acc)
````

# Part 5: Comparison (Relational) & Identity Operators


Comparison operators evaluate the relative ordering or equality between operands and yield a `bool`.

### 1. Relational Operators:
- `==`: Equality (compares contents/values).
- `!=`: Inequality.
- `<`, `<=`: Strictly less than, less than or equal to.
- `>`, `>=`: Strictly greater than, greater than or equal to.

### 2. Chained Comparisons:
Python permits chaining comparison operators, mirroring standard mathematical notation:
- `0 <= score <= 100` evaluates as `(0 <= score) and (score <= 100)`.

### 3. Equality (`==`) vs. Identity (`is`):
- `a == b`: Checks whether `a` and `b` hold equivalent values.
- `a is b`: Checks whether `a` and `b` reference the **exact same memory address** (`id(a) == id(b)`).


````python
# 5.1 Relational Comparisons and Chaining (For Beginners)

# --- What are Relational Operators? ---
# Relational operators are used to compare two values. They always return
# either True or False (a boolean value).
# Common relational operators include: == (equal to), != (not equal to),
# < (less than), > (greater than), <= (less than or equal to),
# >= (greater than or equal to).

# Example 1: Basic Comparisons
num1 = 10
num2 = 20

print(f"Is {num1} equal to {num2}? {num1 == num2}")       # Checks if values are the same
print(f"Is {num1} not equal to {num2}? {num1 != num2}")  # Checks if values are different
print(f"Is {num1} less than {num2}? {num1 < num2}")       # Checks if num1 is smaller than num2
print(f"Is {num1} greater than {num2}? {num1 > num2}")    # Checks if num1 is larger than num2

# Example 2: Chained Comparisons (Pythonic way to check if a value is within a range)
# Python allows you to 'chain' comparison operators like in mathematics.
# For example, `min <= value <= max` is a concise way to check if `value` is
# between `min` and `max` (inclusive).

temp = 24.5 # Let's say this is the current room temperature

# We want to check if the temperature is comfortable, which is between 18.0°C and 26.0°C.
# This single line checks: (18.0 <= temp) AND (temp <= 26.0)
is_comfortable = 18.0 <= temp <= 26.0

print(f"\nIs {temp}°C comfortable (between 18°C and 26°C)? {is_comfortable}")

# Let's try another temperature outside the range
other_temp = 15.0
is_comfortable_other = 18.0 <= other_temp <= 26.0
print(f"Is {other_temp}°C comfortable (between 18°C and 26°C)? {is_comfortable_other}")


# Example 3: Comparing Characters and Strings (Lexicographical Order)
# When you compare strings, Python compares them character by character,
# based on their ASCII/Unicode values. This is called 'lexicographical' order.
# Uppercase letters have lower ASCII values than lowercase letters.

print("\n--- String Comparisons ---")

# 'apple' comes before 'banana' alphabetically, so it's 'less than'.
print("'apple' < 'banana' :", 'apple' < 'banana')

# 'A' (ASCII 65) has a smaller value than 'a' (ASCII 97).
# So, 'A' is considered 'less than' 'a'.
print("'A' < 'a'           :", 'A' < 'a')

# String comparisons are case-sensitive. 'cat' and 'Cat' are different.
print("'cat' == 'Cat'       :", 'cat' == 'Cat')

# To make a case-insensitive comparison, you often convert both strings to the same case:
str1 = "Python"
str2 = "python"
print(f"Is '{str1}' equal to '{str2}' (case-sensitive)? {str1 == str2}")
print(f"Is '{str1}' equal to '{str2}' (case-insensitive)? {str1.lower() == str2.lower()}")
````

````python
# 5.2 Equality (==) vs Identity (is)

# --- Understanding '==' (Equality Operator) and 'is' (Identity Operator) ---
# In Python, we often want to compare things. There are two main ways to compare:
# 1. '==' (Equality): Checks if the *values* of two variables are the same.
#    Think of it as: "Do they have the same content or numerical value?"
# 2. 'is' (Identity): Checks if two variables refer to the *exact same object in memory*.
#    Think of it as: "Are they literally the same thing, stored in the same place?"
#    For simple numbers, Python sometimes optimizes by making identical values
#    point to the same object (called "interning"). This can make 'is' behave
#    like '==' for small numbers, but the core meaning of 'is' remains identity.

print("--- Comparing Simple Numbers ---")

# Example 1: Using '==' for value comparison with user input
# Let's get some numbers from the user.
print("\n--- Input Numbers for Comparison ---")
num_str_a = input("Enter your first number: ")
num_str_b = input("Enter your second number: ")

# Convert the string inputs to integers so we can compare their values.
num_a = int(num_str_a)
num_b = int(num_str_b)

print(f"\nYour first number is: {num_a}")
print(f"Your second number is: {num_b}")

# Check if their *values* are the same using '=='.
# This is usually what you want when comparing numbers.
are_values_equal = (num_a == num_b)
print(f"Are the values of {num_a} and {num_b} equal (using ==)? {are_values_equal}")

# Now let's see if they are the *same exact object* in memory using 'is'.
# For numbers, this might be True for small numbers due to Python's internal optimization.
are_same_object = (num_a is num_b)
print(f"Are {num_a} and {num_b} the *exact same object* (using is)? {are_same_object}")
print("(Note: For small numbers, Python often stores identical values as the same object.)")

## ADvnaced beginners can skip this in lab if they dont understand

# Example 2: Demonstrating Identity with an assigned variable
print("\n--- Demonstrating 'is' for True Identity ---")
my_number = 100
another_variable = my_number # 'another_variable' now points to the *exact same object* as 'my_number'
a_different_number = 100    # This might or might not be the *exact same object* as 'my_number'

print(f"my_number: {my_number}, Memory ID: {hex(id(my_number))}")
print(f"another_variable: {another_variable}, Memory ID: {hex(id(another_variable))}")
print(f"a_different_number: {a_different_number}, Memory ID: {hex(id(a_different_number))}")

# Check value equality
print(f"my_number == another_variable: {my_number == another_variable} (Values are the same)")
print(f"my_number == a_different_number: {my_number == a_different_number} (Values are the same)")

# Check object identity
# 'my_number is another_variable' should always be True because we assigned one to the other.
print(f"my_number is another_variable: {my_number is another_variable} (They point to the exact same object)")
# 'my_number is a_different_number' might be True or False depending on number size and Python's interning.
# The *concept* of 'is' is about the memory ID, regardless of interning.
print(f"my_number is a_different_number: {my_number is a_different_number} (May vary for numbers due to optimization)")


# Example 3: Special Case - Comparing with 'None'
# 'None' is a special constant in Python that represents the absence of a value.
# It's a best practice to *always* use 'is' or 'is not' when checking if a variable is None.
# This is because 'None' is a singleton object, meaning there's only one 'None' object
# in Python's memory, so all variables assigned 'None' will refer to that *exact same object*.
val_none = None
val_zero = 0

print(f"\n--- Special Case: Comparing with 'None' ---")
print(f"Is 'val_none' None? {val_none is None} (Always use 'is' for None!)")
print(f"Is 'val_zero' None? {val_zero is None} (0 is not None)")

# Using '==' with None works, but 'is' is preferred for clarity and consistency,
# and also because 'is' is generally faster for singleton checks.
print(f"Is 'val_none' == None? {val_none == None} (Works, but 'is' is the Pythonic way)")
````

---# Part 6: Logical Operators & Short-Circuit Evaluation

Imagine you're making decisions in daily life. Sometimes you need to combine different conditions. For example, "Is it sunny *AND* warm?" or "Do I have enough money *OR* do I have a discount code?". In programming, **Logical Operators** help us combine these kinds of conditions.

They work by looking at things that are either `True` or `False`. But Python is extra smart: it can also decide if something is "empty" or "not empty" (we call this 'truthiness' and 'falsiness') and use that to make decisions.

### The Three Logical Operators

There are three main logical operators in Python:

1.  **`not` (The 'Opposite' Operator):**
    *   This operator simply flips a `True` into `False` and a `False` into `True`.
    *   **Example:**
        *   `not True` gives you `False`.
        *   `not False` gives you `True`.

2.  **`and` (The 'Both Must Be True' Operator):**
    *   The `and` operator only says `True` if *BOTH* the condition on its left *AND* the condition on its right are `True`.
    *   If even one side is `False`, the whole thing is `False`.
    *   **Example:**
        *   `(2 > 1) and (5 < 10)` is `True` because both `(2 > 1)` (which is `True`) and `(5 < 10)` (which is `True`) are true.
        *   `(2 > 1) and (5 > 10)` is `False` because even though `(2 > 1)` is `True`, `(5 > 10)` is `False`.

3.  **`or` (The 'At Least One Must Be True' Operator):**
    *   The `or` operator is more lenient. It says `True` if *AT LEAST ONE* of the conditions (left or right) is `True`.
    *   It only says `False` if *BOTH* sides are `False`.
    *   **Example:**
        *   `(2 > 1) or (5 > 10)` is `True` because `(2 > 1)` is `True` (even though `(5 > 10)` is `False`).
        *   `(2 < 1) or (5 > 10)` is `False` because both `(2 < 1)` (which is `False`) and `(5 > 10)` (also `False`) are false.

### "Truthiness" and "Falsiness" (More Than Just True/False)

In Python, many types of data can be treated as if they were `True` or `False` in certain situations (like in an `if` statement or with logical operators). This is a really powerful concept!

*   **"Falsy" values** (things Python treats as `False`):
    *   The explicit boolean `False`.
    *   `None` (which means "no value" or "nothing").
    *   The number `0` (zero) in any form (integer `0`, float `0.0`).
    *   Empty text: `""` (an empty string).
    *   Empty lists: `[]` (like an empty shopping list).
    *   Other empty collections like `()` (empty tuples) or `{}` (empty dictionaries).

*   **"Truthy" values** (things Python treats as `True`):
    *   The explicit boolean `True`.
    *   Any number that is *not* `0` (e.g., `1`, `-5`, `3.14`).
    *   Any text that is *not* empty (e.g., `"hello"`, `"a"`, `" "`).
    *   Any list that contains at least one item (e.g., `[1, 2]`, `['apple']`).
    *   Most other objects that are not explicitly defined as falsy.

### What `and` and `or` ACTUALLY Return (A Key Python Trick!)

This is a super important and unique feature of Python's logical operators. Instead of *always* giving back `True` or `False`, `and` and `or` return *one of the actual values they are comparing*.

Let's break down how this works:

#### For `and`:
*   If the **first value (`x`) is "falsy"**, the `and` operator knows the result must be falsy (because `True and False` is `False`, and `False and anything` is `False`). So, it **stops immediately** and returns that **first falsy value (`x`)**.
*   If the **first value (`x`) is "truthy"**, the `and` operator has to check the second value (`y`) to know the overall result. In this case, it evaluates `y` and **returns the second value (`y`)**.

```python
print(0 and 5)      # Output: 0. Because 0 is falsy, 'and' returns 0.
print(10 and 20)    # Output: 20. Because 10 is truthy, 'and' evaluates and returns 20.
print("" and "hello") # Output: "". Because "" (empty string) is falsy, 'and' returns "".
print("hi" and "hello") # Output: "hello". Because "hi" is truthy, 'and' evaluates and returns "hello".
```

#### For `or`:
*   If the **first value (`x`) is "truthy"**, the `or` operator knows the result must be truthy (because `True or anything` is `True`). So, it **stops immediately** and returns that **first truthy value (`x`)**.
*   If the **first value (`x`) is "falsy"**, the `or` operator has to check the second value (`y`) to know the overall result. In this case, it evaluates `y` and **returns the second value (`y`)**.

```python
print(10 or 20)     # Output: 10. Because 10 is truthy, 'or' returns 10.
print(0 or 5)       # Output: 5. Because 0 is falsy, 'or' evaluates and returns 5.
print("hello" or "hi") # Output: "hello". Because "hello" is truthy, 'or' returns "hello".
print("" or "hi")    # Output: "hi". Because "" is falsy, 'or' evaluates and returns "hi".
```

### Short-Circuit Evaluation (Python Being Smart!)

This behavior, where `and` and `or` stop evaluating as soon as they know the final answer, is called **Short-Circuit Evaluation**.

It's not just a cool trick; it's very useful! Python saves time by not doing unnecessary calculations. More importantly, it can help you avoid errors.

Consider this example: trying to divide by zero would normally crash your program. But with short-circuiting, you can prevent it!

```python
num = 10
denom = 0

# Here's the trick:
# 1. Python first checks `(denom != 0)`. Since `denom` is 0, this is `False`.
# 2. Because the first part of an `and` statement is `False`, Python immediately knows the whole expression `False and (anything)` will be `False`.
# 3. It **short-circuits**! It doesn't even try to calculate `(num / denom)`, which would cause a 'ZeroDivisionError'.
safe_division_check = (denom != 0) and (num / denom > 2.5)
print(f"Safe division check result: {safe_division_check}") # Will print False, no error occurs!


num = 10
denom = 5

# In this case:
# 1. `(denom != 0)` is `True`.
# 2. Since the first part is `True`, `and` needs to check the second part. So, `(num / denom > 2.5)` is evaluated.
# 3. `(10 / 5 > 2.5)` becomes `(2.0 > 2.5)`, which is `False`.
safe_division_check = (denom != 0) and (num / denom > 2.5)
print(f"Safe division check result: {safe_division_check}") # Will print False
```
This **defensive guard pattern** using short-circuiting is very common in Python to write robust code that avoids common errors!

````python
# 6.1 Short-Circuit Demonstration with Helper Functions
##  Beginners can skip this code . 
def side_effect_1():
    print("   [Triggered] side_effect_1() executed")
    return False

def side_effect_2():
    print("   [Triggered] side_effect_2() executed")
    return True

print("--- Testing: side_effect_1() and side_effect_2() ---")
res1 = side_effect_1() and side_effect_2()
print("Outcome:", res1)
print("Notice: side_effect_2 was NEVER called!\n")

print("--- Testing: side_effect_2() or side_effect_1() ---")
res2 = side_effect_2() or side_effect_1()
print("Outcome:", res2)
print("Notice: side_effect_1 was NEVER called!")
````

# Part 7: Operator Precedence, Associativity & Formula Translation


When an expression contains multiple mixed operators, Python evaluates them using strict mathematical hierarchy.

### The Complete Precedence Table:
| Rank | Category | Operators | Associativity |
| :---: | :--- | :---: | :---: |
| **1 (Highest)** | Parentheses (Grouping) | `()` | Innermost to Outermost |
| **2** | Exponentiation | `**` | **Right-to-Left** |
| **3** | Unary Operators | `+x`, `-x`, `~x` | Right-to-Left |
| **4** | Multiplicative Operators | `*`, `/`, `//`, `%` | Left-to-Right |
| **5** | Additive Operators | `+`, `-` | Left-to-Right |
| **6** | Bitwise Shifts | `<<`, `>>` | Left-to-Right |
| **7** | Bitwise AND, XOR, OR | `&`, `^`, `\|` | Left-to-Right |
| **8** | Comparisons & Membership | `==`, `!=`, `<`, `<=`, `>`, `>=`, `is`, `in` | Left-to-Right |
| **9** | Logical NOT | `not` | Right-to-Left |
| **10** | Logical AND | `and` | Left-to-Right |
| **11** | Logical OR | `or` | Left-to-Right |
| **12 (Lowest)**| Assignment | `=`, `+=`, `-=`, etc. | Right-to-Left |

### Best Practice Tip for Engineers & Data Scientists:
> **"Explicit is better than implicit."** (The Zen of Python).  
> When writing complex mathematical expressions, use parentheses `()` freely. They cost nothing in performance and completely eliminate operator ambiguity for fellow researchers and code reviewers!


````python
# 7.1 Tracing Step-by-Step Expression Evaluation
# Consider: 15 + 3 * 4 ** 2 // 6 - 8

# Step 1: Exponentiation (4 ** 2 = 16)
# Expression: 15 + 3 * 16 // 6 - 8

# Step 2: Multiplication & Floor Div left-to-right (3 * 16 = 48 -> 48 // 6 = 8)
# Expression: 15 + 8 - 8

# Step 3: Addition and Subtraction left-to-right (15 + 8 = 23 -> 23 - 8 = 15)
calculated = 15 + 3 * 4 ** 2 // 6 - 8
print("Result of 15 + 3 * 4 ** 2 // 6 - 8 =", calculated)
````

````python
# 7.2 Translating Real-World Formulas into Python Expressions

# Formula 1: Gravitational Force between two celestial bodies
# F = G * (m1 * m2) / r^2
G = 6.67430e-11   # Gravitational constant (m^3 kg^-1 s^-2)
mass_earth = 5.972e24  # kg
mass_moon = 7.342e22   # kg
distance = 3.844e8     # meters

F_gravity = G * (mass_earth * mass_moon) / (distance ** 2)
print(f"Earth-Moon Gravitational Force: {F_gravity:.3e} N")

# Formula 2: Electrical Resistors in Parallel
# 1/R_eq = 1/R1 + 1/R2 + 1/R3  ->  R_eq = 1 / (1/R1 + 1/R2 + 1/R3)
r1, r2, r3 = 100.0, 220.0, 470.0
r_eq = 1.0 / ( (1.0 / r1) + (1.0 / r2) + (1.0 / r3) )
print(f"Parallel Equivalent Resistance: {r_eq:.2f} Ohms")
````
