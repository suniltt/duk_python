### 🌟 OOP Explained

Imagine you are a lead designer at a **Smartphone Company**.

Instead of building every single phone from scratch without plans, your team creates a master engineering design first.

* **Class (The Factory Blueprint)**: This is the schematics and design of the phone (e.g., "Model X"). It dictates that all Model X phones must have a screen size, a color, a battery capacity, and a power button. It is a concept on paper, not a physical phone you can make calls with.
* **Object (The Physical Phone in your Pocket)**: This is the actual physical device packaged in a box. When you buy the phone, you have instantiated the "Model X" design into a real object that occupies space on your desk.
* **Attributes (State / Properties)**: These are the features of your individual phone (e.g., `color = "Space Gray"`, `storage_gb = 128`, `battery_charge = 85`).
* **Methods (Behavior / Actions)**: These are the functions built into the phone that perform actions (e.g., `make_call()`, `take_photo()`, `charge_battery()`).

#### 🎨

```text
      +-----------------------------------------+
      |            CLASS: PhoneBlueprint        | <--- The Design/Template
      |-----------------------------------------|
      |  Attributes:                            |
      |    - color                              |
      |    - storage_gb                         |
      |  Methods:                               |
      |    - make_call(number)                  |
      +-----------------------------------------+
                           |
             +-------------+------------+
             | Manufacturing (Instantiating)
             v                          v
  +--------------------+      +--------------------+
  |  OBJECT: my_phone  |      | OBJECT: your_phone | <--- The Actual Objects
  |--------------------|      |--------------------|
  |  color = "Silver"  |      |  color = "Black"   |
  |  storage_gb = 256  |      |  storage_gb = 128  |
  +--------------------+      +--------------------+
```

---

### 🛡️ The 3 Core Pillars of OOP

To make our smartphone factory robust, we follow three main guidelines:

#### 1. Encapsulation
When you use a phone, you don't manually wire the circuit board to make a call. You press a button, and the internal complexity is hidden from you under a sleek protective outer casing.
* In code, **Encapsulation** means hiding the internal working details of an object and only exposing a clean interface. It prevents outside users from accidentally breaking internal components.

#### 2. Inheritance
Next year, you want to build "Model X Pro". Instead of drawing a brand-new blueprint from scratch, you inherit the original "Model X" design and just add a triple-lens camera.
* In code, **Inheritance** allows a new class (subclass) to inherit attributes and methods from an existing class (base class), avoiding code duplication.

#### 3. Polymorphism (
No matter which phone you own—an entry-level model, a Pro model, or even a tablet—they all have a power button. You press it, and they turn on. How each device boots up internally might differ (tablets load a larger UI), but the action you perform is identical.
* In code, **Polymorphism** (meaning "many forms") allows us to trigger the same action (like `power_on()`) across different objects, and each object responds in its own specialized way.

# Part 1: Procedural vs. Object-Oriented Paradigms & The `self` Parameter

Until now, we have practiced **Procedural Programming** where data and logic are separated.

### 🔄 Procedural Approach (Separate Data & Functions)
```python
# Data is stored in loose variables
drone_model = "Surveyor-V1"
drone_x, drone_y = 10.0, 20.0
drone_battery = 100.0

# Independent functions process the data
def fly_to(x, y, target_x, target_y, battery):
    if battery > 10:
        return target_x, target_y, battery - 15.0
    return x, y, battery

drone_x, drone_y, drone_battery = fly_to(drone_x, drone_y, 50.0, 80.0, drone_battery)
```

### 📦 Object-Oriented Approach: The `Drone` Blueprint
In **Object-Oriented Programming (OOP)**, we package both **state** (attributes) and **behavior** (methods) together inside a single entity: the **Object**.

```
+------------------------------------------------------------------------------------+
|                                 CLASS: Drone                                       |
|------------------------------------------------------------------------------------|
|   Attributes (State / Data)                                                        |
|     - model         (e.g., "Surveyor-V1")                                          |
|     - x, y          (e.g., coordinates 10.0, 20.0)                                 |
|     - battery_level (e.g., 100.0)                                                  |
|------------------------------------------------------------------------------------|
|   Methods (Behavior / Actions)                                                     |
|     - take_off()    -> Boots systems up                                            |
|     - fly_to(x, y)  -> Changes coordinates, drains battery                         |
|     - land()        -> Safe descent                                                |
|     - recharge()    -> Restores battery to 100%                                    |
+------------------------------------------------------------------------------------+
```

---

### 🎓 Core OOP   Concepts

#### 1.   Attributes  
Attributes represent the variables inside an object that store its personal, individual data. Every instance created from the `Drone` class will have its own independent copies of `model`, `x`, `y`, and `battery_level`.

#### 2.   Methods (Behavior)
Methods are functions defined inside a class that operate on the object's attributes. Unlike regular functions, they must include `self` as their very first parameter.

#### 3. Demystifying the `self` Parameter
`self` is a reference placeholder representing **"the specific object currently calling this method"**.

When we instantiate two unique drones:
```python
drone_A = Drone("Surveyor-V1")
drone_B = Drone("Explorer-X")
```
And call a method on one of them:
```python
drone_A.take_off()
```
Python automatically translates this invocation behind the scenes to:
```python
Drone.take_off(drone_A)
```
This ensures that `self` points directly to `drone_A`'s attributes, allowing Python to modify `drone_A`'s state without touching `drone_B`.

````python
# 1.1 Defining a Minimal Class and Instantiating Objects
class Drone:
    # A docstring defines the purpose of the class. It is stored in the __doc__ attribute of the class.
    '''Represents an autonomous aerial surveillance drone.'''

    # An instance method defining drone behavior
    # 'self' is the first parameter of any instance method, pointing to the calling object.
    def beep(self):
        # hex(id(self)) returns the unique memory address of the specific object invoking this method
        print(f"[Drone {hex(id(self))}] System active: BEEP!")

# Instantiating (creating) two concrete, separate drone objects in memory:
# Each call to Drone() allocates a unique block of memory and returns a reference to it.
drone_alpha = Drone()
drone_beta = Drone()

# Displaying the class type and unique memory addresses (references) of the instantiated objects
print("drone_alpha Object Reference :", drone_alpha)
print("drone_beta  Object Reference :", drone_beta)

# Invoking the 'beep' method on each instance:
# Python automatically passes the calling object (e.g. drone_alpha) as the first argument ('self').
drone_alpha.beep()
drone_beta.beep()

# Proving how Python dispatches methods under the hood:
# Calling 'drone_alpha.beep()' is syntactic sugar for calling the class method explicitly
# and manually passing the instance as 'self': Drone.beep(drone_alpha)
Drone.beep(drone_alpha)
````

````python
# =========================================================
#  Instance Variables
# =========================================================

class Drone:
    def __init__(self, model, x=0.0, y=0.0):    # This is a special method called constructor
        # These are instance variables.
        # Every time we create a new Drone, a separate set of these variables is created in memory.
        self.model = model
        self.x = float(x)
        self.y = float(y)

# 1. Create two separate, independent instances (objects)
drone_alpha = Drone(model="Quad-X", x=10.0, y=10.0)
drone_beta = Drone(model="Hexa-Y", x=50.0, y=50.0)

print("--- Initial States ---")
print(f"Drone 1 Model: {drone_alpha.model} | Position: ({drone_alpha.x}, {drone_alpha.y})")
print(f"Drone 2 Model: {drone_beta.model} | Position: ({drone_beta.x}, {drone_beta.y})")

# 2. Modify instance variables of drone_alpha (Quad-X)
print("\n--- Modifying Quad-X position only ---")
drone_alpha.x = 99.0
drone_alpha.y = 99.0

# 3. Observe the result: Changes to drone_alpha do NOT affect drone_beta!
print("\n--- Verification ---")
print(f"Drone 1 Model: {drone_alpha.model} | Position: ({drone_alpha.x}, {drone_alpha.y}) <--- Changed!")
print(f"Drone 2 Model: {drone_beta.model} | Position: ({drone_beta.x}, {drone_beta.y}) <--- Unchanged!")
````

````python
# =========================================================

# =========================================================

class Drone:
    '''Represents an autonomous aerial surveillance drone.'''

    def __init__(self, model, x=0.0, y=0.0, battery_level=100.0):
        # These are instance attributes. Every drone object created has its own personal copy.
        self.model = model                      # Stores the name/id of the drone model
        self.x = float(x)                       # Drone's current X coordinate in space
        self.y = float(y)                       # Drone's current Y coordinate in space
        self.battery_level = float(battery_level) # Battery percentage remaining
        self.is_flying = False                  # Track state: Is the drone airborne (True) or landed (False)?

        print(f"[System] '{self.model}' initialized at ({self.x}, {self.y}) with {self.battery_level}% battery.")

    def take_off(self):
        '''Changes the flying state of the drone if it has enough power.'''
        # We check the object's attribute (self.battery_level) before acting
        if self.battery_level > 10:
            self.is_flying = True
            print(f"[Take-off] {self.model} successfully took off!")
        else:
            print(f"[Error] Cannot take off. Battery level ({self.battery_level}%) is too low!")



# 1. Instantiate two unique drone objects
drone_alpha = Drone(model="Quad-X", x=10.0, y=12.0, battery_level=95.0)
drone_beta = Drone(model="Hexa-Y", x=100.0, y=200.0, battery_level=8.0)

print("\n--- Executing Method: take_off() ---")
# 2. Invoke the 'take_off' method on drone_alpha (This should succeed)
drone_alpha.take_off()

# 3. Invoke the 'take_off' method on drone_beta (This should fail due to low battery safety check)
drone_beta.take_off()
````

````python
# =========================================================

# =========================================================

class Drone:
    '''Represents an autonomous aerial surveillance drone.'''

    # -----------------------------------------------------
    # -----------------------------------------------------
    def __init__(self, model, x=0.0, y=0.0, battery_level=100.0):
        # These are instance attributes. Every drone object created has its own personal copy.
        self.model = model                      # Stores the name/id of the drone model
        self.x = float(x)                       # Drone's current X coordinate in space
        self.y = float(y)                       # Drone's current Y coordinate in space
        self.battery_level = float(battery_level) # Battery percentage remaining
        self.is_flying = False                  # Track state: Is the drone airborne (True) or landed (False)?

        print(f"[System] '{self.model}' initialized at ({self.x}, {self.y}) with {self.battery_level}% battery.")

    # -----------------------------------------------------

    # -----------------------------------------------------
    def take_off(self):
        '''Changes the flying state of the drone if it has enough power.'''
        # We check the object's attribute (self.battery_level) before acting
        if self.battery_level > 10:
            self.is_flying = True
            print(f"[Take-off] {self.model} successfully took off!")
        else:
            print(f"[Error] Cannot take off. Battery level ({self.battery_level}%) is too low!")

    # -----------------------------------------------------

    # -----------------------------------------------------
    def fly_to(self, target_x, target_y):
        '''Moves the drone to a new location, updating coordinates and battery.'''
        # Safety check 1: The drone must be airborne to fly somewhere
        if not self.is_flying:
            print(f"[Error] {self.model} cannot fly. It must take off first!")
            return

        # Safety check 2: The drone needs enough battery to make a trip
        if self.battery_level > 20:
            print(f"[Flight] Moving {self.model} from ({self.x}, {self.y}) to ({target_x}, {target_y})")
            # We update the attributes on 'self' (the active object)
            self.x = float(target_x)
            self.y = float(target_y)
            self.battery_level -= 15.0  # Drains battery power by 15%
            print(f"[Flight] Arrived! New Battery level: {self.battery_level}%")
        else:
            print(f"[Warning] Battery too low ({self.battery_level}%). Safe-landing initiated.")
            self.land()  # We can call another method inside this class using self!

    # -----------------------------------------------------

    # -----------------------------------------------------
    def land(self):
        '''Brings the drone safely back to the ground.'''
        if self.is_flying:
            self.is_flying = False
            print(f"[Landing] {self.model} has landed safely.")
        else:
            print(f"[System] {self.model} is already on the ground.")

    # -----------------------------------------------------
    # -----------------------------------------------------
    def recharge(self):
        '''Resets the battery state back to maximum.'''
        self.battery_level = 100.0
        print(f"[Recharge] {self.model} battery recharged back to 100%.")


print("--- Executing Complete Drone Demo ---")
# 1. Instantiate the drone (Runs __init__)
my_drone = Drone(model="Surveyor-V1", x=10.0, y=20.0)

# 2. Try flying before taking off (Demonstrating safety checks)
my_drone.fly_to(50.0, 80.0)

# 3. Take off and fly successfully (State updates)
my_drone.take_off()
my_drone.fly_to(50.0, 80.0)

# 4. Land and recharge (Completing lifecycle)
my_drone.land()
my_drone.recharge()
````

# Part 2: Class Variables vs. Instance Variables

One of the most critical concepts in Python OOP is the distinction between **Class Variables** and **Instance Variables**.

```
+------------------------------------------------------------------------------------+
|                         Class Namespace: Employee                                  |
|   company_name = "ICFOSS Kerala"  <--- SHARED across all employee instances!       |
|   total_employees = 2             <--- Shared counter incremented in __init__      |
+------------------------------------------------------------------------------------+
            ^                                           ^
            |                                           |
+-----------------------+                   +-----------------------+
|  Instance: emp1       |                   |  Instance: emp2       |
|  name = "Anjana J"    |                   |  name = "Renij R"     |
|  salary = 75000.0     |                   |  salary = 72000.0     |
+-----------------------+                   +-----------------------+
```

### The Rules of Namespace Lookup:
1. **Instance Variables**: Bound to `self.` inside methods. Each object maintains its own independent copy.
2. **Class Variables**: Defined directly inside the `class` block, outside any method. Shared by **all** instances of that class.
3. **The Shadowing Pitfall**: If you execute `emp1.company_name = "New Org"`, Python does **not** update the shared class variable! Instead, it creates a new *instance attribute* on `emp1` that masks/shadows the class variable for `emp1` only!
4. **Best Practice**: Always read and modify class variables using the class name: `Employee.company_name = "New Org"`.

````python
# =========================================================

# =========================================================
# Analogy:
# - Instance Variables: Like properties on your own personal passport (name, birth date).
# - Class Variables: Like your country's passport color (everyone in the country shares it).
# =========================================================

class SmartDrone:
    # 1. CLASS VARIABLES (Shared by ALL objects created from this class)
    fleet_name = "SkyGuard Alpha"
    operating_frequency_ghz = 5.8
    total_active_drones = 0  # Counter that increments as new drones are initialized

    def __init__(self, designation, pilot_name):
        # 2. INSTANCE VARIABLES (Unique to each specific drone object)
        self.designation = designation
        self.pilot_name = pilot_name

        # Increment the shared class counter whenever a new drone is built
        SmartDrone.total_active_drones += 1
        print(f"[Factory] Built '{self.designation}' for pilot {self.pilot_name}.")

    def report_status(self):
        # We access instance variables with 'self' and class variables with the class name (or self)
        print(f"- {self.designation} is on fleet '{SmartDrone.fleet_name}' using {SmartDrone.operating_frequency_ghz} GHz.")


# --- STEP 1: Inspect Class Variables BEFORE creating any objects ---
print("=== Class Variables exist on the Class itself ===")
print("Current Fleet:", SmartDrone.fleet_name)
print("Active Drone Count:", SmartDrone.total_active_drones)
print()

# --- STEP 2: Instantiate two separate drones ---
print("=== Instantiating Objects ===")
drone_1 = SmartDrone("Drone-01", "Alice")
drone_2 = SmartDrone("Drone-02", "Bob")
print()

# --- STEP 3: Verify unique instance variables vs shared class variables ---
print("=== Status Reports ===")
drone_1.report_status()
drone_2.report_status()
print("\nTotal Drones in Fleet (Shared Count):", SmartDrone.total_active_drones)
print()

# --- STEP 4: Modify Class Variable once, affecting all instances ---
print("=== Updating Operating Frequency (Class Variable) ===")
# By updating the class variable, EVERY drone's configuration changes immediately
SmartDrone.operating_frequency_ghz = 2.4

print(f"{drone_1.designation} Frequency: {drone_1.operating_frequency_ghz} GHz")
print(f"{drone_2.designation} Frequency: {drone_2.operating_frequency_ghz} GHz")
````

````python
# 3.1 Class Variables vs. Instance Variables in Action
class AIResearcher:
    # 1. CLASS VARIABLES (Shared across all instances!)
    institute = "Digital University Kerala"
    annual_travel_grant = 150000.0
    total_researchers = 0

    def __init__(self, name, primary_domain):
        # 2. INSTANCE VARIABLES (Unique to each researcher!)
        self.name = name
        self.domain = primary_domain
        self.publications = 0

        # Increment the shared class counter upon every instantiation:
        AIResearcher.total_researchers += 1

# Create researchers:
r1 = AIResearcher("Dr. Arun", "Computer Vision")
r2 = AIResearcher("Dr. Bindu", "Natural Language Processing")

print(f"Researcher 1: {r1.name} | Institute: {r1.institute}")
print(f"Researcher 2: {r2.name} | Institute: {r2.institute}")
print(f"Total Researchers Count (via Class)   : {AIResearcher.total_researchers}")
print(f"Total Researchers Count (via Instance): {r1.total_researchers}")
````

````python
# 3.2 THE SHADOWING PITFALL: Modifying Class Variables Incorrectly!

# Case A: Correct Modification via Class Name:
AIResearcher.annual_travel_grant = 175000.0
print("After updating AIResearcher.annual_travel_grant:")
print("  r1 grant:", r1.annual_travel_grant)  # 175000.0 (Updated!)
print("  r2 grant:", r2.annual_travel_grant)  # 175000.0 (Updated!)

# Case B: The Accidental Shadowing Trap via Instance:
r1.annual_travel_grant = 250000.0  # ACCIDENTAL INSTANCE VARIABLE CREATION!
print("\nAfter writing: r1.annual_travel_grant = 250000.0:")
print("  r1 grant (shadowed attribute):", r1.annual_travel_grant)  # 250000.0
print("  r2 grant (shared class var)  :", r2.annual_travel_grant)  # 175000.0
print("  Class level grant            :", AIResearcher.annual_travel_grant)  # 175000.0

# Inspecting namespaces:
print("\nr1 __dict__:", r1.__dict__)  # Has 'annual_travel_grant' explicitly!
print("r2 __dict__:", r2.__dict__)  # Does NOT have 'annual_travel_grant' (falls back to class!)
````

### Constructors & Instance Initialization (`__init__`)

When an object is created, Python automatically executes a special **dunder (double underscore) method** called `__init__()`.

```
       drone = Drone("Surveyor-V1", 100.0)
                  |
                  v
       1. Python allocates new memory block for object.
       2. Python calls: Drone.__init__(new_object, "Surveyor-V1", 100.0)
       3. Attributes are bound: self.model = "Surveyor-V1", self.battery = 100.0
       4. Initialized object reference is returned and bound to 'drone'.
```

### Magic (Dunder) Methods for String Representation:
- **`__str__(self)`**: Returns an informal, user-friendly string (invoked by `print(obj)` and `str(obj)`).
- **`__repr__(self)`**: Returns an unambiguous, formal string for debugging (invoked in the REPL and `repr(obj)`).

````python
# 2.1 Implementing the Constructor (__init__)
class RobotArm:
    '''Represents an industrial 6-DOF robotic manipulator.'''

    # The Constructor Method: This special dunder method is automatically called
    # whenever a new instance of RobotArm is created.
    # It accepts arguments to initialize the object's unique state.
    def __init__(self, arm_id, payload_capacity_kg, reach_mm=1200):
        # Bind incoming arguments to the instance namespace via 'self.'
        self.arm_id = arm_id
        self.payload_kg = float(payload_capacity_kg)
        self.reach_mm = reach_mm

        # We can also initialize internal state attributes to a default value
        self.is_clamped = False
        print(f"[Constructor] Initialized RobotArm '{self.arm_id}' successfully.")

    def toggle_gripper(self):
        '''Toggles gripper between clamped and open states.'''
        # Modify the instance attribute 'is_clamped' using 'self'
        self.is_clamped = not self.is_clamped
        state_str = "CLAMPED" if self.is_clamped else "OPEN"
        print(f"RobotArm '{self.arm_id}' gripper is now: {state_str}")

# Creating instances with mandatory and optional constructor parameters:
# arm1 uses the default parameter reach_mm=1200
arm1 = RobotArm("ARM-101", 15.0)

# arm2 overrides the default reach_mm with 2500
arm2 = RobotArm("ARM-102", 50.0, reach_mm=2500)

# Printing and verifying each object's distinct attributes
print(f"\nArm 1: {arm1.arm_id} | Payload: {arm1.payload_kg} kg | Reach: {arm1.reach_mm} mm")
print(f"Arm 2: {arm2.arm_id} | Payload: {arm2.payload_kg} kg | Reach: {arm2.reach_mm} mm")

# Manipulating state on 'arm1' using its instance methods
arm1.toggle_gripper()
arm1.toggle_gripper()
````

````python
# 2.2 Magic Methods for Human-Readable Inspection: __str__ and __repr__
class ScientificPoint:
    '''Represents a 2D coordinate point with physics units.'''
    def __init__(self, x, y, unit="meters"):
        self.x = float(x)
        self.y = float(y)
        self.unit = unit

    # Informal string representation: intended for end-users
    def __str__(self):
        return f"({self.x:.2f}, {self.y:.2f}) {self.unit}"

    # Formal string representation: intended for developers & debugging
    def __repr__(self):
        return f"ScientificPoint(x={self.x}, y={self.y}, unit='{self.unit}')"

pt = ScientificPoint(12.456, 8.912)

# Calling str(pt) or print(pt) triggers __str__:
print("Formatted print(pt) :", pt)

# Calling repr(pt) triggers __repr__:
print("Debug repr(pt)      :", repr(pt))
````

### Task C1: The Foundation `Employee` Class & Constructor
**Context**: Human Resources Management Systems require clean object modeling to track corporate staff.

**Problem Statement**:
1. Implement the base `Employee` class with:
   - Constructor `__init__(self, emp_id, name, department, basic_monthly_salary)`.
   - Instance attributes: `self.emp_id`, `self.name`, `self.department`, `self.basic_salary`.
   - Method `calculate_annual_ctc(self, allowance_pct=25.0)`:
     $$\text{Annual CTC} = (\text{basic\_salary} \times 12) \times \left(1 + \frac{\text{allowance\_pct}}{100}\right)$$
   - Method `display_details(self)`: Prints a formatted summary card of the employee.
2. Instantiate two employees (`E101` and `E102`) and compute their annual CTC.


````python
# TASK C1: WORKSPACE


class Employee:
    '''Base class representing an organization employee.'''

    def __init__(self, emp_id, name, department, basic_monthly_salary):
        # TODO: Initialize instance attributes
        self.emp_id = str(emp_id)
        self.name = str(name)
        self.department = str(department)
        self.basic_salary = float(basic_monthly_salary)

    def calculate_annual_ctc(self, allowance_pct=25.0):
        '''Computes total annual CTC with allowances.'''
        # TODO: Return annual salary plus allowances
        annual_base = self.basic_salary * 12
        total_ctc = annual_base * (1.0 + (allowance_pct / 100.0))
        return round(total_ctc, 2)

    def display_details(self):
        '''Displays formatted employee summary.'''
        ctc = self.calculate_annual_ctc()
        print(f"[{self.emp_id}] {self.name:<18} | Dept: {self.department:<14} | Monthly: Rs. {self.basic_salary:>9,.2f} | CTC: Rs. {ctc:>11,.2f}")

# Test instances:
emp1 = Employee("EMP-101", "Arun Kumar", "AI Research", 65000.0)
emp2 = Employee("EMP-102", "Bhavana Nair", "Data Platform", 72000.0)

emp1.display_details()
emp2.display_details()
````

### Task C2: Class Variables for Enterprise Roster Tracking
**Context**: Company-wide payroll and headcount analytics aggregate data across all instances.

**Problem Statement**:
Enhance the `Employee` class by introducing **Class Variables**:
1. `company_name = "ICFOSS Technologies"`
2. `total_employees = 0` (increments by 1 inside `__init__`)
3. `total_monthly_payroll = 0.0` (accumulates `self.basic_salary` in `__init__`)
4. Add a class-level method `get_company_stats(cls)`:
   Returns a dictionary:
   ```python
   {
       "company": cls.company_name,
       "headcount": cls.total_employees,
       "monthly_payroll": cls.total_monthly_payroll,
       "annual_payroll": cls.total_monthly_payroll * 12
   }
   ```


````python
# TASK C2: WORKSPACE
# -------------------------------------------------------------------

class EmployeeV2:
    # 1. CLASS VARIABLES:
    company_name = "ICFOSS Technologies"
    total_employees = 0
    total_monthly_payroll = 0.0

    def __init__(self, emp_id, name, department, basic_monthly_salary):
        self.emp_id = str(emp_id)
        self.name = str(name)
        self.department = str(department)
        self.basic_salary = float(basic_monthly_salary)

        # 2. Update class variables automatically:
        EmployeeV2.total_employees += 1
        EmployeeV2.total_monthly_payroll += self.basic_salary

    @classmethod
    def get_company_stats(cls):
        '''Returns aggregate company-wide payroll statistics.'''
        return {
            "company": cls.company_name,
            "headcount": cls.total_employees,
            "monthly_payroll": cls.total_monthly_payroll,
            "annual_payroll": cls.total_monthly_payroll * 12
        }

# Instantiate employees:
e1 = EmployeeV2("E201", "Chitra S", "Embedded AI", 60000.0)
e2 = EmployeeV2("E202", "Deepak R", "Computer Vision", 80000.0)
e3 = EmployeeV2("E203", "Esha M", "NLP Systems", 75000.0)

stats = EmployeeV2.get_company_stats()
print("=== Company Roster Statistics ===")
for k, v in stats.items():
    if isinstance(v, float):
        print(f"  {k:<18}: Rs. {v:,.2f}")
    else:
        print(f"  {k:<18}: {v}")
````
